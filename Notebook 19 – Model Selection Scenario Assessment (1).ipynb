# Notebook 19 – Model Selection Scenario Assessment

## 1. What Is Model Selection?

Model selection is the process of choosing the most suitable AI/ML model for a particular problem.

Different models are designed for different tasks. For example, a speech recognition model converts audio into text, while an image classification model identifies objects or categories in images.

## 2. What Should We Consider?

- **Problem:** What task must the model solve?
- **Input type:** Is the input text, audio, image, or structured data?
- **Expected output:** What result should the model produce?
- **Suitable model type:** What type of model can solve the task?
- **Candidate models:** Which models should we compare?
- **Selection criteria:** Which model provides the best balance of quality, speed, and cost?
- **Hardware considerations:** Can the model run on available CPU, GPU, and RAM?
- **Limitations:** What errors or weaknesses might the model have?
- **Final justification:** Why is this model the best choice for the given requirements?

## 3. Important Principle

The largest or most accurate model is not always the best choice. A suitable model should meet the task's requirements while considering performance, hardware, cost, and limitations.

```python
scenario = {
    "problem": "Customer sentiment analysis",
    "input_type": "Text",
    "expected_output": "Positive or Negative",
    "suitable_model_type": "Text classification",
    "candidate_models": [
        "DistilBERT",
        "BERT"
    ],
    "selection_criteria": [
        "Accuracy",
        "Inference speed",
        "Memory usage"
    ],
    "hardware": "CPU or GPU",
    "limitations": "May misunderstand sarcasm or context",
    "final_choice": "DistilBERT"
}

for key, value in scenario.items():
    print(f"{key}: {value}")
```

# Scenario 01 – Customer Sentiment Analysis

## 1. Problem
A company receives thousands of customer reviews and wants to automatically identify whether each review expresses a positive, negative, or neutral opinion.

## 2. Input Type
Text data, such as customer reviews, feedback, and product comments.

Example: "The product quality is excellent."

## 3. Expected Output
A sentiment label: Positive, Negative, or Neutral.

Example output: Positive

## 4. Suitable Model Type
A pretrained text classification model fine-tuned for sentiment analysis.

## 5. Candidate Models
- DistilBERT
- BERT
- RoBERTa

## 6. Selection Criteria
- Classification accuracy and F1-score
- Inference speed
- Memory usage
- Ability to identify positive, negative, and neutral sentiment
- Performance on the company's actual reviews

## 7. Hardware Considerations
DistilBERT can be a practical choice for CPU inference when traffic and latency requirements are moderate. A GPU may help when processing large volumes of reviews. Actual resource requirements depend on the model configuration and workload.

## 8. Limitations
- May misunderstand sarcasm, such as "Excellent service — I waited three hours."
- May struggle with slang, spelling errors, or mixed sentiments.
- Performance may decrease on reviews that differ from its training data.

## 9. Final Justification
DistilBERT is a suitable initial choice because it is smaller and generally faster than BERT while retaining useful text classification capabilities. It is appropriate when inference speed and resource efficiency matter. The final choice should be validated using representative company reviews and the required performance metrics.

```python
scenario_1 = {
    "problem": "Customer sentiment analysis",
    "input_type": "Customer review text",
    "expected_output": ["Positive", "Negative", "Neutral"],
    "suitable_model_type": "Text classification",
    "candidate_models": ["DistilBERT", "BERT", "RoBERTa"],
    "selection_criteria": ["F1-score", "Speed", "Memory"],
    "hardware": "CPU or GPU",
    "limitations": [
        "Sarcasm",
        "Slang",
        "Mixed sentiments"
    ],
    "final_choice": "DistilBERT"
}

for section, details in scenario_1.items():
    print(f"{section}: {details}")
```

# Scenario 02 – Call Transcription

## 1. Problem
A company wants to convert customer service call recordings into text for documentation, searching, and analysis.

## 2. Input Type
Audio data, such as WAV, MP3, or recorded telephone conversations.

## 3. Expected Output
Text containing the spoken words from the audio recording.

Example input: A customer says, "I need help tracking my order."

Example output: "I need help tracking my order."

## 4. Suitable Model Type
Automatic Speech Recognition (ASR), also called a speech-to-text model.

## 5. Candidate Models
- Whisper
- Wav2Vec2
- NVIDIA NeMo ASR models

## 6. Selection Criteria
- Word Error Rate (WER)
- Recognition of accents and different speaking styles
- Performance with background noise
- Transcription speed
- Support for required languages
- Ability to process long recordings

## 7. Hardware Considerations
Smaller ASR models can be suitable for CPU inference, although transcription may take longer. A GPU can accelerate processing when supported. Larger models generally require more memory and computing resources.

## 8. Limitations
- Background noise can cause incorrect transcription.
- Accents and overlapping speakers may reduce accuracy.
- Technical terms, names, and regional languages may be misrecognized.
- Speech-to-text models can produce incorrect words that sound similar to the original speech.

## 9. Final Justification
Whisper is a suitable candidate because it supports multilingual speech recognition and can transcribe recorded audio. A smaller Whisper model can be evaluated for CPU-based workloads, while larger models may provide better recognition quality at higher computational cost. The final choice should be based on measured WER, speed, language requirements, and hardware availability.

```python
scenario_2 = {
    "problem": "Transcribe customer service calls",
    "input_type": "Audio",
    "expected_output": "Recognized text",
    "suitable_model_type": "Automatic Speech Recognition",
    "candidate_models": [
        "Whisper",
        "Wav2Vec2",
        "NVIDIA NeMo ASR"
    ],
    "selection_criteria": [
        "Word Error Rate",
        "Inference speed",
        "Language support",
        "Noise handling"
    ],
    "hardware": "CPU or GPU",
    "limitations": [
        "Background noise",
        "Accents",
        "Overlapping speech"
    ],
    "final_choice": "Whisper"
}

for section, details in scenario_2.items():
    print(f"{section}: {details}")
```

# Scenario 03 – Document Entity Extraction

## 1. Problem
A company wants to automatically extract important information from documents, such as invoices, contracts, resumes, and business reports.

## 2. Input Type
Text extracted from documents, or document images processed using OCR (Optical Character Recognition).

Example input:
"John joined ABC Technologies on 15 March 2025."

## 3. Expected Output
Identified entities with their categories.

Example output:
- Person: John
- Organization: ABC Technologies
- Date: 15 March 2025

## 4. Suitable Model Type
Named Entity Recognition (NER) model.

NER identifies specific entities in text and classifies them into categories such as person, organization, location, and date.

## 5. Candidate Models
- BERT fine-tuned for NER
- spaCy pretrained NER models
- RoBERTa fine-tuned for NER

For documents containing complex layouts, layout-aware models such as LayoutLM can also be considered.

## 6. Selection Criteria
- Entity extraction precision, recall, and F1-score
- Accuracy on company-specific document types
- Support for required entity categories
- Processing speed
- Ability to handle document layout and formatting

## 7. Hardware Considerations
A smaller NER model can run on a CPU for moderate workloads. A GPU may improve throughput when processing large document collections. OCR and layout-aware models may require additional resources.

## 8. Limitations
- May fail to recognize unfamiliar names or company-specific terminology.
- OCR errors can affect extraction accuracy.
- A model trained on general text may perform poorly on specialized documents.
- Standard NER may identify entities without understanding the full meaning of a contract or invoice.

## 9. Final Justification
A pretrained BERT model fine-tuned for NER is a suitable candidate when the goal is to extract names, organizations, dates, and other defined entities from text. It can be adapted to company-specific categories using labeled examples. If document layout is important, a layout-aware model such as LayoutLM may be more appropriate. The final choice should be validated against manually labeled documents.

```python
scenario_3 = {
    "problem": "Extract entities from business documents",
    "input_type": "Document text or OCR text",
    "expected_output": [
        "Person",
        "Organization",
        "Date"
    ],
    "suitable_model_type": "Named Entity Recognition",
    "candidate_models": [
        "BERT for NER",
        "spaCy NER",
        "RoBERTa for NER",
        "LayoutLM"
    ],
    "selection_criteria": [
        "Precision",
        "Recall",
        "F1-score",
        "Processing speed"
    ],
    "hardware": "CPU or GPU",
    "limitations": [
        "OCR errors",
        "Unfamiliar entities",
        "Specialized terminology"
    ],
    "final_choice": "BERT fine-tuned for NER"
}

for section, details in scenario_3.items():
    print(f"{section}: {details}")
```

# Scenario 04 – Image Classification

## 1. Problem
A manufacturing company wants to automatically classify product images to identify whether a product is normal or defective.

## 2. Input Type
Digital images, usually represented as RGB pixel arrays.

Example input: An image of a manufactured screw.

## 3. Expected Output
A predicted class label, such as:
- Normal
- Defective

The model may also return confidence scores for the predicted classes.

## 4. Suitable Model Type
Image classification model based on a Convolutional Neural Network (CNN) or Vision Transformer (ViT).

## 5. Candidate Models
- ResNet-18
- MobileNetV3
- EfficientNet
- Vision Transformer (ViT)

## 6. Selection Criteria
- Classification accuracy and F1-score
- Ability to detect different defect types
- Inference speed
- Model size and memory usage
- Performance on unseen product images
- Cost of incorrect defect predictions

## 7. Hardware Considerations
MobileNetV3 is a reasonable candidate for CPU inference and resource-constrained environments. ResNet-18 can serve as a practical baseline. A GPU can accelerate training and high-volume inference. Actual requirements depend on image resolution and workload.

## 8. Limitations
- Poor lighting and blurry images can reduce accuracy.
- Defects not represented in training data may be missed.
- Class imbalance can cause the model to overlook rare defects.
- A classification model identifies a class but does not necessarily show the defect's exact location.

## 9. Final Justification
MobileNetV3 is a suitable starting candidate when fast inference and low resource usage are important. ResNet-18 can be used as a baseline for comparison. The selected model must be tested on representative factory images, especially rare defects. If the application needs defect locations or pixel-level regions, object detection or segmentation may be more appropriate than image classification.

```python
scenario_4 = {
    "problem": "Classify manufactured products",
    "input_type": "RGB images",
    "expected_output": ["Normal", "Defective"],
    "suitable_model_type": "Image classification",
    "candidate_models": [
        "ResNet-18",
        "MobileNetV3",
        "EfficientNet",
        "Vision Transformer"
    ],
    "selection_criteria": [
        "Accuracy",
        "F1-score",
        "Inference speed",
        "Model size"
    ],
    "hardware": "CPU or GPU",
    "limitations": [
        "Poor lighting",
        "Unseen defects",
        "Class imbalance",
        "No defect localization"
    ],
    "final_choice": "MobileNetV3"
}

for section, details in scenario_4.items():
    print(f"{section}: {details}")
```

# Scenario 05 – Enterprise Chatbot

## 1. Problem
A company wants to build an internal chatbot that answers employee questions about company policies, IT support procedures, HR documents, and technical documentation.

## 2. Input Type
Natural language text, such as employee questions and company documents.

Example input: "How do I reset my company password?"

## 3. Expected Output
A relevant, understandable answer based on approved company information.

Example output: Instructions for resetting the company password, based on the official IT documentation.

## 4. Suitable Model Type
A Large Language Model (LLM) combined with Retrieval-Augmented Generation (RAG).

RAG retrieves relevant information from company documents and provides it to the language model to generate an answer.

## 5. Candidate Models
- Llama
- Mistral
- GPT models
- Smaller instruction-tuned language models

The exact model depends on deployment requirements, privacy policies, licensing, and available hardware.

## 6. Selection Criteria
- Answer accuracy and relevance
- Ability to follow instructions
- Hallucination rate
- Response speed
- Context window size
- Data privacy and security
- Deployment cost
- Quality of answers grounded in company documents

## 7. Hardware Considerations
A smaller language model may run locally with suitable RAM or GPU memory. Larger models generally need more computing resources. Hosted models can reduce local hardware requirements, but introduce service costs and data-handling considerations.

## 8. Limitations
- The model may generate incorrect or unsupported answers.
- Outdated or incomplete company documents can produce poor responses.
- Sensitive information must be protected through access controls.
- RAG does not guarantee factual answers.
- Large models can be expensive or slow to operate.

## 9. Final Justification
An instruction-tuned LLM with RAG is a suitable approach because the chatbot needs to understand employee questions and generate answers grounded in company documentation. RAG can help keep answers aligned with updated documents without retraining the language model for every document change. The solution should be evaluated for answer quality, retrieval accuracy, security, latency, and cost.
