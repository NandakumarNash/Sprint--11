# Topic 1: Model Repository

## Simple Concept

A **model repository** is a website or platform where we can find pre-trained models, read their details, and download or use them.

Instead of training a model from the beginning, we can choose an existing model from a repository.

## Important Explanation

A model repository helps us check:

- Model name and architecture
- Supported tasks
- Input and output formats
- Model size and number of parameters
- Training dataset
- License and usage restrictions
- Hardware requirements
- Model limitations

Different repositories provide models for different purposes, such as image classification, object detection, text generation, and speech recognition.

## Examples of Model Repositories

1. **Hugging Face** – Provides models for computer vision, natural language processing, and audio.
2. **PyTorch Hub** – Provides models that can be used with PyTorch.
3. **TorchVision** – Provides pre-trained computer vision models, such as ResNet and MobileNet.

## Example

Suppose we want to classify images of cats and dogs.

Instead of building a CNN from scratch, we can visit a model repository, find a pre-trained image classification model, check its requirements, and decide whether it fits our task.

**Key point:** A model repository helps us discover and evaluate models before choosing one for our project.

# Topic 3: Model Architecture

## Simple Concept

**Model architecture** describes how a model is structured and how its layers or components work together to process input and produce output.

Different architectures are designed for different tasks.

## Important Explanation

In computer vision, common model architectures include:

- **CNN:** Extracts visual features from images.
- **ResNet:** Uses skip connections to help train deeper neural networks.
- **MobileNet:** Designed to use fewer computational resources, making it suitable for mobile and lightweight applications.
- **Vision Transformer (ViT):** Uses attention mechanisms to understand relationships between image patches.

When selecting a pre-trained model, we should consider whether its architecture is suitable for our task and available hardware.

For example, if we need image classification on a resource-limited device, MobileNet may be a suitable choice. If we need a deeper architecture, ResNet may be worth evaluating.


````python

## Python Example

from torchvision import models

resnet_model = models.resnet18(
    weights=models.ResNet18_Weights.DEFAULT
)

mobilenet_model = models.mobilenet_v2(
    weights=models.MobileNet_V2_Weights.DEFAULT
)

print("ResNet architecture:")
print(resnet_model)

print("\nMobileNet architecture:")
print(mobilenet_model)
````

```

This code loads two pre-trained architectures and displays their layer structures. The printed output is long because each model contains many layers.

**Key point:** Choose an architecture based on the task, accuracy requirements, and computational resources—not just its name.

# Topic 4: Model Size

## Simple Concept

**Model size** refers to the amount of storage space required to save a model's weights and related files.

Model size is usually measured in MB or GB.

## Important Explanation

Different pre-trained models require different amounts of storage.

- Smaller models are generally easier to download, store, and deploy on devices with limited resources.
- Larger models may contain more parameters and may offer greater capabilities, but they can require more storage and computing resources.

Model size depends on factors such as the number of parameters, their numerical precision, and additional files included in the model.

For example, a model saved using 32-bit floating-point weights will generally require more storage than the same weights saved using 16-bit precision.


````python

## Python Example

from torchvision import models
import torch

model = models.mobilenet_v2(
    weights=models.MobileNet_V2_Weights.DEFAULT
)

torch.save(model.state_dict(), "mobilenet_v2.pth")

size_mb = (
    __import__("os").path.getsize("mobilenet_v2.pth")
    / (1024 * 1024)
)

print(f"Model size: {size_mb:.2f} MB")
````

```

**Expected output:** The program displays the saved model's size in MB. The exact value may vary depending on the model weights and saved format.

**Key point:** Check model size before choosing a model, especially when storage space or deployment-device capacity is limited.

# Topic 5: Number of Parameters

## Simple Concept

**Parameters** are the values a model learns during training, such as weights and biases.

The number of parameters tells us how many learnable values a model contains.

## Important Explanation

Pre-trained models can have different numbers of parameters.

- Models with fewer parameters are often smaller and more efficient.
- Models with more parameters may be able to learn more complex patterns, but they can require more memory and computation.
- More parameters do not automatically mean better performance. The task, training data, and model architecture also matter.

For example, MobileNetV2 has approximately 3.5 million parameters, while ResNet-50 has approximately 25.6 million parameters.
.

````python

## Python Example

from torchvision import models

mobilenet = models.mobilenet_v2(
    weights=models.MobileNet_V2_Weights.DEFAULT
)

resnet = models.resnet50(
    weights=models.ResNet50_Weights.DEFAULT
)

mobilenet_params = sum(
    p.numel() for p in mobilenet.parameters()
)

resnet_params = sum(
    p.numel() for p in resnet.parameters()
)

print("MobileNetV2 parameters:", f"{mobilenet_params:,}")
print("ResNet-50 parameters:", f"{resnet_params:,}")
````

# Topic 7: Output Format

## Simple Concept

**Output format** describes the type and structure of information a model produces after processing its input.

Different models produce different types of outputs depending on their tasks.

## Important Explanation

Examples of model outputs:

- **Image classification:** Class scores or probabilities for categories such as cat, dog, and car.
- **Object detection:** Object classes, bounding-box coordinates, and confidence scores.
- **Image segmentation:** A pixel-level mask identifying the regions belonging to different classes.
- **Text generation:** Generated text.
- **Speech recognition:** Transcribed text from audio.

For example, a pre-trained ResNet-18 model returns a tensor containing 1,000 class scores when used with its standard ImageNet classification head.

These scores are not probabilities by default. We can apply softmax to convert them into probabilities.


````python

## Python Example

import torch
from torchvision import models

weights = models.ResNet18_Weights.DEFAULT
model = models.resnet18(weights=weights)
model.eval()

# Example input: one RGB image, 224 × 224 pixels
image = torch.rand(1, 3, 224, 224)

with torch.no_grad():
    output = model(image)

probabilities = torch.softmax(output, dim=1)

predicted_index = probabilities.argmax(dim=1).item()
class_name = weights.meta["categories"][predicted_index]

print("Output shape:", output.shape)
print("Predicted class:", class_name)
print("Confidence:", round(probabilities[0, predicted_index].item(), 4))
````

# Topic 8: Supported Languages

## Simple Concept

**Supported languages** refers to the human languages a pre-trained model can understand or process.

This topic is especially important for Natural Language Processing (NLP), speech recognition, and text generation models.

## Important Explanation

Different models support different languages.

- **English-only models:** Primarily designed for English text.
- **Multilingual models:** Designed to process multiple languages.
- **Language-specific models:** Designed for a particular language or a smaller group of languages.

For example:

- BERT (English version) is commonly used for English-language tasks.
- mBERT (Multilingual BERT) supports multiple languages.
- XLM-RoBERTa supports multilingual text processing.

When selecting a model, check whether it supports the language used in your input data. A model trained mainly on English may not perform well on Tamil or other languages.

For computer vision models, human-language support is usually not a major selection factor unless the model also processes text, such as an image-text model.


````python

## Python Example

from transformers import AutoTokenizer

model_name = "bert-base-multilingual-cased"

tokenizer = AutoTokenizer.from_pretrained(model_name)

text = "வணக்கம், how are you?"

tokens = tokenizer.tokenize(text)

print("Tokens:", tokens)
````

```

**Expected output:** The tokenizer displays tokens representing the Tamil and English text. The exact tokenization depends on the tokenizer vocabulary.

This example demonstrates text tokenization; it does not prove that the model performs equally well in every supported language.

**Key point:** Select a model whose language support matches your project's actual language requirements.

# Topic 10: Model Performance

## Simple Concept

**Model performance** describes how well a pre-trained model performs a particular task.

A model should be evaluated using metrics that match the problem we want to solve.

## Important Explanation

Different tasks require different performance metrics.

- **Image classification:** Accuracy, precision, recall, and F1-score.
- **Object detection:** Mean Average Precision (mAP).
- **Image segmentation:** Intersection over Union (IoU) and mean IoU.
- **Text generation:** Metrics depend on the task, such as BLEU for some translation evaluations.

When comparing pre-trained models, check:

- Performance on relevant test datasets.
- Inference speed.
- Memory consumption.
- Performance on data similar to your project.
- Whether reported results are independently reproducible.

A model with high accuracy on ImageNet may not perform well on industrial defect images because the data and task are different.



````python
## Python Example

from sklearn.metrics import accuracy_score

actual = ["cat", "dog", "cat", "bird", "dog"]
predicted = ["cat", "dog", "dog", "bird", "dog"]

accuracy = accuracy_score(actual, predicted)

print("Accuracy:", accuracy)
````

# Topic 12: Hardware Requirements

## Simple Concept

**Hardware requirements** describe the computer resources needed to load and run a pre-trained model.

These resources may include RAM, storage, processor capability, and GPU memory.

## Important Explanation

Different models require different amounts of computing resources.

- **RAM:** Holds model data and other information while the program runs.
- **Storage:** Stores downloaded model weights and related files.
- **CPU:** Processes model operations using the computer's processor.
- **GPU:** Can accelerate many deep learning operations, especially for large models and batches of images.

For example, MobileNetV2 is relatively lightweight and may be suitable for a computer with limited resources. A much larger model may require more RAM or GPU memory.

Actual requirements also depend on image size, batch size, numerical precision, and whether we are training or performing inference.


````python

## Python Example

import torch
from torchvision import models

model = models.mobilenet_v2(
    weights=models.MobileNet_V2_Weights.DEFAULT
)

parameter_count = sum(
    parameter.numel()
    for parameter in model.parameters()
)

parameter_memory_mb = sum(
    parameter.numel() * parameter.element_size()
    for parameter in model.parameters()
) / (1024 * 1024)

print("Parameters:", f"{parameter_count:,}")
print("Parameter memory:", f"{parameter_memory_mb:.2f} MB")
print("CUDA GPU available:", torch.cuda.is_available())
````

```

**Expected output:** The code displays the parameter count, approximate memory occupied by model parameters, and whether a CUDA-compatible GPU is available.

The parameter memory does not include all runtime memory requirements, such as intermediate activations and other program overhead.

**Key point:** Check whether your computer has enough resources to run the model comfortably before selecting it.

# Topic 13: CPU/GPU Requirements

## Simple Concept

A **CPU** and a **GPU** are processors that can execute a pre-trained model.

- **CPU (Central Processing Unit):** Can run deep learning models without a dedicated graphics processor.
- **GPU (Graphics Processing Unit):** Can accelerate many deep learning operations through parallel processing.

## Important Explanation

When selecting a pre-trained model, consider where you will run it.

- **CPU:** Often sufficient for small models and occasional predictions.
- **GPU:** Often preferable for larger models, high-volume predictions, and faster training or fine-tuning.
- **GPU memory:** Must be sufficient for the model and its runtime workload.

A GPU is not mandatory for every pre-trained model. For example, MobileNetV2 can perform image classification on a CPU, although inference may be slower than on a suitable GPU.

Also, having a GPU does not automatically mean PyTorch can use it. Your installed PyTorch version and hardware must support the required GPU backend.


````python

## Python Example

import torch

if torch.cuda.is_available():
    print("GPU available:", torch.cuda.get_device_name(0))
else:
    print("GPU not available. CPU can be used.")
````

```

**Expected output if CUDA is unavailable:**

```text
GPU not available. CPU can be used.
```

Your output depends on your computer and PyTorch installation.

**Key point:** Choose a model that runs efficiently on the hardware available for your project.

# Topic 14: Model Limitations

## Simple Concept

**Model limitations** are the weaknesses or restrictions that affect how well a pre-trained model performs.

No model performs perfectly in every situation.

## Important Explanation

Common model limitations include:

- **Limited training data:** The model may not recognize patterns absent from its training data.
- **Domain differences:** A model trained on everyday photographs may struggle with industrial defect images.
- **Bias:** The model may perform differently across certain data groups or conditions.
- **Input restrictions:** The model may require a specific image size, format, or input type.
- **Computational cost:** Larger models may be too slow for some devices.
- **Incorrect predictions:** A model can produce wrong classifications even when its confidence is high.

For example, a pre-trained ImageNet classifier may recognize a rabbit in a photograph but fail to identify a small manufacturing defect because it was not specifically trained for that task.

Understanding limitations helps us decide whether to use a model directly, fine-tune it, or select another model.



````python
## Python Example

from torchvision import models

weights = models.ResNet18_Weights.DEFAULT
model = models.resnet18(weights=weights)

print("Number of output classes:",
      len(weights.meta["categories"]))

print("Example supported classes:",
      weights.meta["categories"][:5])
````

**Key point:** Understand a model's limitations before deciding whether it is suitable for your project.


# Topic 16: Out-of-Scope Use

## Simple Concept

**Out-of-scope use** refers to using a pre-trained model for purposes that fall outside its documented capabilities, intended applications, or usage restrictions.

A model may work well for its original task but perform poorly when used for a different purpose.

## Important Explanation

Consider a pre-trained ResNet-18 model trained on ImageNet.

- **Within scope:** Classifying everyday images into the categories supported by its classification head.
- **Potentially out of scope:** Directly using the standard ImageNet model to identify specific manufacturing defects without adapting or evaluating it.
- **Technical limitation:** The standard model does not directly return object bounding boxes or pixel-level segmentation masks.
- **Usage restriction:** A model's license or documentation may prohibit or restrict certain applications.

Out-of-scope use does not always mean that the model can never be adapted for that task. Fine-tuning, changing the model architecture, or training on a suitable dataset may make it useful for a new purpose, subject to applicable restrictions.


````python

## Python Example

from torchvision.models import (
    resnet18,
    ResNet18_Weights
)

weights = ResNet18_Weights.DEFAULT
model = resnet18(weights=weights)

print("Output classes:",
      len(weights.meta["categories"]))

print("Model task: Image classification")
print("Direct object localization: Not supported")
print("Direct pixel segmentation: Not supported")
````

```

```text
Output classes: 1000
Model task: Image classification
Direct object localization: Not supported
Direct pixel segmentation: Not supported
```

These limitations refer to the standard ResNet-18 classification model, not every possible ResNet-based architecture.

**Key point:** Before selecting a model, identify what it can do, what it cannot do directly, and whether your intended use is permitted.
