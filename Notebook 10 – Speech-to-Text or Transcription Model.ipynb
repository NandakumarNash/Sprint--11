### 1. What is Speech-to-Text?

Speech-to-Text (STT) is an AI task that converts spoken words in an audio recording into written text.

For example:

**Audio input:** A person says, "Good morning, how are you?"

**Text output:** `Good morning, how are you?`

### 2. Why is Speech-to-Text used?

- Transcribing meetings and interviews
- Generating subtitles for videos
- Converting voice messages into text
- Supporting voice assistants
- Creating searchable records of conversations

### 3. How does it work?

1. **Audio input:** Provide a recorded voice file.
2. **Audio preprocessing:** Prepare the audio in the format required by the model.
3. **Model inference:** The pre-trained model processes the audio.
4. **Transcription:** The model predicts the spoken words.
5. **Output text:** Display the recognized text.

### 4. Important point

Speech recognition does not always produce the exact words spoken. Background noise, unclear pronunciation, different accents, and speaking speed can affect the result.

**Remember:** Speech-to-Text converts audio into text. The model predicts the words; it does not guarantee a perfect transcription.

# Topic 04 – Sampling Rate

### 1. What is a sampling rate?

Sampling rate is the number of times per second an audio signal is measured to represent sound digitally.

It is measured in Hertz (Hz).

For example:

- **16,000 Hz (16 kHz):** 16,000 audio samples per second.
- **44,100 Hz (44.1 kHz):** 44,100 audio samples per second.

A higher sampling rate represents more detail in the audio signal's frequency range, but it also requires more data.

### 2. Why is sampling rate important?

A speech recognition model expects audio in a suitable format and sampling rate.

Whisper typically processes audio at **16,000 Hz**. If the original recording uses another sampling rate, the audio can be resampled before processing.

**Important:** Changing the sampling-rate label alone is not resampling. The audio data must actually be converted.

### 3. Sampling rate versus audio duration

For a mono recording:

- Sampling rate: 16,000 Hz
- Duration: 2 seconds
- Number of audio samples: 32,000

The number of samples depends on the sampling rate and recording duration.

```python
%pip install soundfile
```

```python
import soundfile as sf

for audio_path in [sample1, sample2]:
    audio, sampling_rate = sf.read(audio_path)

    duration = len(audio) / sampling_rate

    print("File:", audio_path.name)
    print("Sampling rate:", sampling_rate, "Hz")
    print("Duration:", round(duration, 2), "seconds")
    print("Number of samples:", len(audio))
    print()
```

# Topic 05 – Audio Preprocessing

### 1. What is audio preprocessing?

Audio preprocessing means preparing an audio recording before passing it to a speech recognition model.

The audio file may need to be converted into a suitable sampling rate, channel format, or numerical representation.

### 2. Why is preprocessing required?

Audio recordings can differ in:
- Sampling rate
- Number of channels
- Audio format
- Duration and recording quality

The model expects audio in a supported format. Preprocessing helps make the input suitable for inference.

### 3. Common audio preprocessing steps

**Step 1 – Load audio:** Read the audio file into Python.

**Step 2 – Check sampling rate:** Identify how many audio samples are recorded per second.

**Step 3 – Check audio channels:** Determine whether the audio is mono or stereo.

**Step 4 – Resample if necessary:** Convert the audio to the required sampling rate.

**Step 5 – Prepare the audio:** Pass the processed recording to the model.

### 4. Mono versus stereo

- **Mono:** One audio channel.
- **Stereo:** Two audio channels, commonly left and right.

A speech recognition workflow may convert stereo audio to mono when required.

### 5. Important point

Do not change the sampling rate or audio channels unnecessarily. First inspect the recording, then apply only the preprocessing that is needed.

For our Whisper experiment, the model's audio processing workflow handles standard audio preparation internally when we use its normal transcription interface.

```python
import soundfile as sf
import numpy as np

def prepare_audio(audio_path):
    audio, sampling_rate = sf.read(audio_path)

    # Convert stereo audio to mono if needed
    if audio.ndim == 2:
        audio = np.mean(audio, axis=1)

    audio = audio.astype(np.float32)

    print("File:", audio_path.name)
    print("Sampling rate:", sampling_rate, "Hz")
    print("Audio shape:", audio.shape)
    print("Mono audio:", audio.ndim == 1)
    print()

    return audio, sampling_rate

audio1, sr1 = prepare_audio(sample1)
audio2, sr2 = prepare_audio(sample2)
```

# Notebook 10 – Topic 11: Analyzing Transcription Accuracy

## What Is Transcription Accuracy?

Transcription accuracy describes how closely the predicted text matches the words actually spoken in the audio.

For example:

- **Actual speech:** I am learning machine learning.
- **Predicted text:** I am learning machine learning.
- **Result:** The words match correctly.

If the model misses, replaces, or adds words, the transcription contains errors.

## What Is Word Error Rate (WER)?

Word Error Rate (WER) measures transcription errors by comparing the predicted words with the reference transcription.

It considers three types of errors:

- **Substitution:** A word is replaced with another word.
- **Deletion:** A spoken word is missing from the prediction.
- **Insertion:** An extra word appears in the prediction.

A lower WER indicates fewer word-level errors.

## Important Point

A transcription that looks correct is not necessarily perfect. We should compare it against the actual spoken words.

WER can also be greater than 100% when there are many insertion errors.

```python
# Reference text: the words actually spoken
reference1 = "Good morning. I am learning speech recognition and machine learning."
reference2 = "Today is a sunny day. I want to convert this audio recording into text."

# Predicted text from Whisper
print("Reference 1:", reference1)
print("Predicted 1:", result1["text"])

print("\nReference 2:", reference2)
print("Predicted 2:", result2["text"])
```

# Notebook 10 – Topic 12: Calculating Word Error Rate (WER)

## What Is Word Error Rate?

Word Error Rate (WER) measures how many word-level errors appear in a transcription compared with the reference text.

It considers three types of errors:

- **Substitution (S):** A word is replaced by an incorrect word.
- **Deletion (D):** A spoken word is missing.
- **Insertion (I):** An extra word is added.

## Formula

WER is calculated as:

WER = (S + D + I) / N

Here, N is the total number of words in the reference transcription.

## Example

Reference: `I like machine learning`

Prediction: `I love machine learning`

The word `like` was replaced by `love`, so there is one substitution.

There are 4 words in the reference and prediction, with 1 substitution.

WER = 1 / 4 = 0.25 = 25%

A lower WER indicates a closer match. WER can exceed 100% if there are many insertion errors.

```python
%pip install jiwer
```

```python
from jiwer import wer

# Actual spoken sentences
reference1 = "Good morning. I am learning speech recognition and machine learning."
reference2 = "Today is a sunny day. I want to convert this audio recording into text."

# Calculate WER
wer1 = wer(reference1, result1["text"])
wer2 = wer(reference2, result2["text"])

print(f"Sample 1 WER: {wer1:.2%}")
print(f"Sample 2 WER: {wer2:.2%}")
```

# Notebook 10 – Topic 13: How Pronunciation Affects Transcription

## What Is Pronunciation?

Pronunciation is the way a person produces spoken words.

Speech recognition models analyze audio patterns to predict words. If words are spoken unclearly or unusually, the model may predict them incorrectly.

## How Pronunciation Affects Results

- **Clear pronunciation:** Words may be easier for the model to recognize.
- **Unclear pronunciation:** Some words may be missed or replaced.
- **Different accents:** Pronunciation differences may affect recognition.
- **Fast speech:** Connected words can be harder to distinguish.

## Key Point

Clear pronunciation can help speech recognition, but even clearly spoken audio can produce errors. The model's training data, audio quality, and background noise also influence its predictions.

# Notebook 10 – Topic 14: Effect of Background Noise

## What Is Background Noise?

Background noise is any unwanted sound in an audio recording, such as traffic, fans, music, or people talking nearby.

## How Background Noise Affects Speech Recognition

- **Clean audio:** Speech is easier for the model to identify.
- **Low background noise:** The model may still recognize most words correctly.
- **High background noise:** Speech can become difficult to distinguish, leading to transcription errors.

## Why Does This Happen?

Speech recognition models process audio patterns to predict words. Background noise can interfere with the speech information needed to make accurate predictions.

## Key Point

Audio quality can affect transcription accuracy. However, the impact depends on the type and level of noise and the model being used.

```python
import numpy as np
import soundfile as sf
import whisper

# Load audio
audio, sr = sf.read(str(sample1))

# Convert stereo audio to mono if necessary
if audio.ndim == 2:
    audio = audio.mean(axis=1)

# Add low-level artificial noise
rng = np.random.default_rng(42)
noise = rng.normal(0, 0.01, size=audio.shape)
noisy_audio = audio + noise

# Save noisy audio
noisy_path = audio_folder / "sample1_noisy.wav"
sf.write(str(noisy_path), noisy_audio, sr)

# Transcribe noisy audio
noisy_result = model.transcribe(str(noisy_path), fp16=False)

print("Original:", result1["text"])
print("Noisy:", noisy_result["text"])
```

# Notebook 10 – Topic 15: Effect of Different Speakers and Speaking Speeds

## 1. Different Speakers

People have different voices, accents, pitches, and pronunciation styles.

A speech recognition model may produce different results for different speakers, even when they say the same sentence.

## 2. Different Speaking Speeds

- **Slow speech:** Words may be easier to distinguish, although unnaturally slow speech can also affect recognition.
- **Normal speech:** The model processes speech at its usual pace.
- **Fast speech:** Words may blend together, potentially causing transcription errors.

## 3. Why Test These Conditions?

Testing different voices and speaking speeds helps us understand how well a model handles variations in real speech.

## Key Point

Speaker characteristics and speaking speed can affect transcription results. To evaluate them fairly, compare recordings of the same sentence under different conditions.

```python
import soundfile as sf
import numpy as np

# Load the original audio
audio, sr = sf.read(str(sample1))

# Create a slower version by changing playback speed
slow_audio = audio[::2]

slow_path = audio_folder / "sample1_slow.wav"

# Save at the original sample rate
sf.write(str(slow_path), slow_audio, sr)

# Transcribe the slower audio
slow_result = model.transcribe(str(slow_path), fp16=False)

print("Original transcription:")
print(result1["text"])

print("\nSlower audio transcription:")
print(slow_result["text"])
```
