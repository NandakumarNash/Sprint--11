# Topic 1: What Is Hugging Face?

Hugging Face is a platform that helps developers and learners find, use, share, and work with pre-trained machine learning models.

It provides ready-made models for tasks such as:

- Text classification
- Text generation
- Image classification
- Object detection
- Speech recognition
- Translation

Instead of building and training every model from the beginning, we can use a pre-trained model from Hugging Face.

**Example:**

Suppose we want to classify a customer review as positive or negative.

We can search for a pre-trained sentiment analysis model on Hugging Face and use it to predict the sentiment of the review.

**Important points:**

- Hugging Face provides access to pre-trained models and datasets.
- It supports different machine learning tasks.
- It provides libraries and tools for loading and using models.
- Many resources are free to access, although some models and services may have restrictions or costs.

**Important point:**

Hugging Face is the overall platform, while the Hugging Face Hub is its central place for discovering and sharing models, datasets, and applications.

# Topic 4: Datasets on Hugging Face Hub

A dataset is a collection of data used to train, validate, or test a machine learning model.

The Hugging Face Hub provides datasets for different machine learning tasks, including text classification, language translation, image classification, and speech recognition.

**Types of datasets**

- **Text datasets:** Contain sentences, documents, or reviews.
- **Image datasets:** Contain images and sometimes their labels.
- **Audio datasets:** Contain audio recordings and related information.
- **Tabular datasets:** Contain structured data arranged in rows and columns.

**Example:**

Suppose we want to build a sentiment analysis model that classifies customer reviews as positive or negative.

We can find a suitable review dataset on Hugging Face Hub. The dataset may contain review text and sentiment labels.

**How datasets are used**

1. Collect or select a suitable dataset.
2. Prepare and clean the data.
3. Use training data to train the model.
4. Use validation data to tune the model when needed.
5. Use test data to evaluate the model.

**Important point:**

A dataset is different from a model. A dataset contains the examples, while a model learns patterns from those examples and produces predictions.

# Topic 6: Model Repositories

A model repository is a dedicated location on Hugging Face Hub that stores the files and information related to a particular model.

Each model repository has its own page and repository ID.

**Example:**

A repository ID might look like this:

`google-bert/bert-base-uncased`

This identifies a particular model repository on Hugging Face Hub.

**What does a model repository contain?**

- **Model weights:** The learned values used by the model.
- **Configuration files:** Information needed to configure the model architecture.
- **Tokenizer files:** Files required to convert text into tokens for compatible language models.
- **Model card:** Documentation explaining the model's purpose, usage, and limitations.
- **Additional files:** Other files required for model use or provided by the author.

Not every repository contains exactly the same files. The contents depend on the model and how its author has packaged it.

**Why are model repositories useful?**

1. They keep a model's files and documentation together.
2. They help users inspect a model before using it.
3. They provide a location from which compatible tools can download model files.
4. They may include version history and information about updates.

**Important point:**

A model repository is not just the model weights. It is the complete collection of files and documentation associated with that model.

# Topic 7: Model Cards

A model card is a documentation page that explains a machine learning model, its intended purpose, how to use it, and its limitations.

On Hugging Face Hub, the model card is usually displayed on the model's repository page and is commonly written in a file named `README.md`.

**What information does a model card contain?**

- **Model description:** What the model does.
- **Intended use:** The tasks and situations the model is designed for.
- **Supported tasks:** The types of predictions or outputs it can produce.
- **Training information:** Details about the training data and process, when provided.
- **Usage instructions:** Examples showing how to load and use the model.
- **Limitations:** Known weaknesses or situations where the model may perform poorly.
- **License:** Conditions governing how the model can be used, when specified.

**Example:**

Suppose we find a sentiment analysis model on Hugging Face Hub.

Before using it, we read its model card to check whether it supports our language, how to provide input text, what output it returns, and whether its license permits our intended use.

**Why are model cards important?**

They help us decide whether a model is suitable for our project before downloading or using it.

**Important point:**

A model card is documentation, not the model itself. The information provided may be incomplete, so we should verify important requirements rather than assume every detail is documented.

# Topic 8: Dataset Cards

A dataset card is a documentation page that explains a dataset, what it contains, how it was collected, and how it should be used.

On Hugging Face Hub, dataset cards are commonly written in a file named `README.md` within the dataset repository.

**What information does a dataset card contain?**

- **Dataset description:** What the dataset contains.
- **Data source:** Where the data came from, when known.
- **Dataset structure:** The columns, features, or types of data available.
- **Data splits:** Information about training, validation, and test sets, when provided.
- **Intended use:** The tasks the dataset is suitable for.
- **Limitations and bias:** Possible problems or limitations in the data.
- **License:** Conditions for using or sharing the dataset, when specified.

**Example:**

Suppose we want to build a sentiment analysis model.

Before downloading a customer review dataset, we read its dataset card to check whether it contains review text and sentiment labels, whether the data is suitable for our task, and whether we are permitted to use it.

**Why are dataset cards important?**

They help us understand the dataset before using it for training or evaluation.

**Important point:**

A dataset card describes the dataset; it does not contain the model's learned parameters. Always check the actual dataset files and usage conditions as well.

# Topic 9: Model Files

Model files are the files stored in a model repository on Hugging Face Hub. They contain the learned parameters, architecture configuration, and other resources needed to use a model.

**Common model files**

- **`model.safetensors`**: Stores model weights in the SafeTensors format.
- **`pytorch_model.bin`**: Commonly stores model weights in a PyTorch checkpoint format.
- **`config.json`**: Contains configuration information about the model architecture.
- **`tokenizer.json`**: Stores tokenizer information for compatible models.
- **`tokenizer_config.json`**: Contains tokenizer configuration details.
- **`vocab.txt`**: Contains vocabulary entries for tokenizers that use this format.
- **`README.md`**: Usually contains the model card and usage documentation.

Not every model repository contains all these files. File names and formats depend on the model and the framework used.

**Example:**

Suppose we want to load a pre-trained text classification model.

We may need its model weights, configuration, and tokenizer files. The loading library uses these files to reconstruct the model and prepare text for prediction.

**Why should we inspect model files?**

1. To identify the files required to load a model.
2. To understand which framework or format is being used.
3. To check whether the necessary files are available.
4. To avoid downloading unrelated files unnecessarily.

**Important point:**

Model weights contain learned numerical values. Configuration files describe how the model should be constructed. They serve different purposes.

# Topic 10: Configuration Files

A configuration file contains settings and information that describe how a machine learning model should be constructed or loaded.

On Hugging Face Hub, a common model configuration file is `config.json`.

**What information can a model configuration contain?**

- **Model architecture:** The type of model to construct.
- **Number of layers:** The number of layers used in the architecture, when applicable.
- **Hidden size:** The size of internal representations in certain neural networks.
- **Number of attention heads:** The number of attention heads in Transformer models.
- **Number of labels:** The number of output classes for some classification models.
- **Vocabulary size:** The number of tokens supported by a model's vocabulary, where applicable.

The exact settings depend on the model architecture.

**Example:**

A text classification model may use a configuration file to specify its architecture and the number of output labels.

When we load the model, the library reads the configuration to construct the appropriate model structure before loading its learned weights.

**Why are configuration files important?**

1. They describe how the model should be constructed.
2. They help loading libraries interpret the model files correctly.
3. They reduce the need to manually define the architecture.

**Important point:**

A configuration file does not usually contain the model's learned weights. Those are stored separately in model weight files.



# Topic 11: Model Checkpoints

A model checkpoint is a saved state of a machine learning model at a particular point during training or after training.

It allows us to save a model's learned weights so we can load them later without training the model again from the beginning.

**Why are checkpoints useful?**

- **Resume training:** Continue training from a previously saved state.
- **Save progress:** Keep a copy of the model during training.
- **Compare models:** Compare checkpoints saved at different training stages.
- **Reuse a trained model:** Load saved weights for inference.

**Example:**

Suppose we train an image classification model for 20 epochs.

We save a checkpoint after epoch 10 and another after epoch 20. Later, we can load either checkpoint and evaluate its predictions.

**Checkpoint vs. model weights**

- **Model weights:** The learned numerical parameters of the model.
- **Checkpoint:** A saved state that may contain model weights and additional information, such as optimizer state and training progress.

A checkpoint does not always contain all this information; its contents depend on how it was saved.

**Important point:**

Pre-trained models on Hugging Face Hub commonly provide saved model weights that can be loaded directly. A checkpoint saved during training may also contain information needed to resume training.

# Topic 12: Search for Models on Hugging Face Hub

Hugging Face Hub allows us to search for pre-trained models using a model name, task, or keyword.

We can search through the website or use Python libraries to access models.

## Method 1: Search using the website

1. Open https://huggingface.co/models
2. Enter a keyword, such as `image classification`.
3. Browse the available models.
4. Open a model repository to inspect its details.

## Method 2: Search using Python

We can use the `huggingface_hub` library to search for models programmatically.



```python
# Install once if the library is missing
# %pip install -U huggingface_hub

from huggingface_hub import HfApi

api = HfApi()

models = api.list_models(
    search="image classification",
    limit=5
)

for model in models:
    print(model.id)
```

# Topic 13: Filter Models on Hugging Face Hub

Filtering means narrowing search results to find models that match our requirements.

Hugging Face Hub provides filters that help us select models based on their task, library, dataset, license, and other available properties.

**Common filters**

- **Task:** Image classification, text generation, object detection, and more.
- **Library:** Transformers, Diffusers, and other supported libraries.
- **Dataset:** Models associated with a particular training dataset.
- **License:** Models with specified usage permissions.
- **Other filters:** Depending on the Hub interface, options may include model size, language, and other attributes.

**Example:**

Suppose we need a model to classify weather images.

1. Search for `weather image classification`.
2. Select the image classification task filter, if available.
3. Review the matching models.
4. Open a model repository and inspect its model card and requirements.

**Important point:**

Filters help us find suitable candidates, but they do not guarantee that a model is accurate or suitable for our project. We must inspect the model before choosing it.

```python
from huggingface_hub import HfApi

api = HfApi()

models = api.list_models(
    pipeline_tag="image-classification",
    limit=5
)

for model in models:
    print(model.id)
```

# Topic 14 – Identify Task Type on Hugging Face Hub

**What is a task type?**

A task type tells us what a model is designed to do.

For example:
- **Image Classification:** Identifies the category of an image.
- **Object Detection:** Finds objects and their locations in an image.
- **Text Classification:** Assigns a category to text.
- **Text Generation:** Generates text.
- **Image Segmentation:** Identifies objects or regions at the pixel level.

**How to identify a model’s task type**

1. Open the model’s page on Hugging Face Hub.
2. Check the task tag displayed near the model name.
3. Read the model card to understand what the model does.
4. Check the labels and configuration to understand its expected output.

**Example:** The `Weather-Image-Classification` model classifies weather images into categories such as cloudy, foggy, rainy, snowy, and sunny.

**Key point:** Always identify the task type before loading a model. It helps you choose a model suitable for your project.

```python
from huggingface_hub import HfApi

api = HfApi()

model = api.model_info(
    "prithivMLmods/Weather-Image-Classification"
)

print("Model ID:", model.id)
print("Task type:", model.pipeline_tag)
```

# Topic 16 – Understand Model Requirements

**What are model requirements?**

Model requirements are the conditions needed to load and use a model correctly.

Before using a model, check these details:

1. **Input type:** Does the model accept images, text, audio, or another type of input?
2. **Input size:** Does the image need to be resized to a particular size?
3. **Dependencies:** Which Python libraries are required?
4. **Model architecture:** Which architecture does the model use?
5. **Hardware:** Can it run on a CPU, or does it require a GPU for practical performance?
6. **Preprocessing:** Does the input need resizing, normalization, or tokenization?

**Example: Weather image classification**

For the weather classification model we inspected:
- Input: A weather image.
- Image size in the configuration: 224 × 224 pixels.
- Image channels: 3, representing a color image.
- Architecture: SigLIP-based image classification.

The model’s processor and usage instructions determine the exact preprocessing needed.

**Key point:** Check model requirements before loading a model to avoid compatibility errors and incorrect predictions.

```python
from transformers import AutoConfig

model_name = "prithivMLmods/Weather-Image-Classification"

config = AutoConfig.from_pretrained(model_name)

print("Model type:", config.model_type)
print("Number of labels:", config.num_labels)
print("Image size:", config.vision_config.image_size)
print("Number of channels:", config.vision_config.num_channels)
```

# Topic 17 – Download and Load a Model

**What does downloading and loading a model mean?**

- **Download:** Obtain the model files from Hugging Face Hub. These files may be cached on your computer.
- **Load:** Read the model configuration and weights into Python so the model can be used.

**How it works**

1. Choose a suitable model on Hugging Face Hub.
2. Check its task, requirements, and usage instructions.
3. Load the model using the appropriate Python library.
4. Provide the required input.
5. Get the model’s prediction.

**Example:** We can load the weather image classification model and use it to predict the weather category in an image.

**Important:** The first load may take time because the model files need to be downloaded. Internet access is usually required the first time, unless the files are already cached locally.

```python
from transformers import AutoImageProcessor, AutoModelForImageClassification

model_name = "prithivMLmods/Weather-Image-Classification"

processor = AutoImageProcessor.from_pretrained(model_name)
model = AutoModelForImageClassification.from_pretrained(model_name)

print("Model loaded successfully!")
print("Labels:", model.config.id2label)
```
