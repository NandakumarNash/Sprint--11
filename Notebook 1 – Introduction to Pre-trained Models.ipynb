### 1. What is a Model?

A **model** is a computer program that learns patterns from data and uses those patterns to make predictions on new data.

For example, an image classification model can learn the visual differences between cats and dogs.

During training, the model learns patterns such as shapes, edges, ears, and other visual features. When a new image is given, it uses the learned patterns to predict whether the image contains a cat or a dog.

**Simple example:**

- Input: An image of a cat
- Model: A trained image classification model
- Output: Cat

A model does not simply memorize a fixed answer for every possible image. It learns patterns that help it make predictions on images it has not seen before.

**Key point:** A model learns patterns from data and uses them to produce predictions.

```python
image_name = "cat.jpg"

predicted_class = "Cat"

print("Input Image:", image_name)
print("Predicted Class:", predicted_class)
```

## 2. What is a Pre-trained Model?

A **pre-trained model** is a model that has already been trained on a dataset and has learned useful patterns from that data.

Instead of training a model from the beginning, we can use an existing trained model for our task.

**Example in Computer Vision:**

Imagine a model that has already been trained on millions of images.

During training, it may have learned to recognize:
- Edges and lines
- Shapes and textures
- Objects such as cars, dogs, and birds

We can use this model as a starting point for a new task, such as identifying different types of vehicles.

**How it works:**

1. A model is trained using a large dataset.
2. It learns patterns and stores them in its model weights.
3. The trained model is saved.
4. We load the pre-trained model for our task.
5. We give it a new image and obtain a prediction.

**Example:**

- Original training task: Recognize many types of objects.
- New task: Identify different types of cars.
- Approach: Reuse the pre-trained model and adapt it to the car classification task if needed.

A pre-trained model does not automatically solve every new problem. Its usefulness depends on how similar the original training data is to our new task.

**Key point:** A pre-trained model has already learned patterns from previous training, so we can reuse that knowledge instead of starting from scratch.

```python
from torchvision import models

# Load a pre-trained image classification model
model = models.resnet18(weights="DEFAULT")

# Set the model to inference mode
model.eval()

print("Pre-trained model loaded successfully")
```

## 3. Why Use Pre-trained Models?

Pre-trained models are useful because training a model from the beginning can require a large dataset, significant computing power, and considerable time.

A pre-trained model already contains knowledge learned from its original training data. We can reuse that knowledge for our task.

**Main reasons to use pre-trained models:**

1. **Less training time:** We do not always need to train the entire model from the beginning.
2. **Less data required:** A pre-trained model can often learn a new task using a smaller dataset.
3. **Reduced computing requirements:** Reusing learned features can reduce the amount of training computation needed.
4. **Good starting point:** The model already knows useful patterns, such as edges, shapes, and textures in images.
5. **Useful for transfer learning:** We can adapt an existing model to a different but related task.

**Computer Vision example:**

Suppose you want to build a model that identifies defective products in factory images.

Training a deep neural network from scratch may require many labeled images and substantial computing resources.

Instead, you could start with a pre-trained image model, adapt it to your product images, and train it to recognize defective and non-defective products.

**Important:** A pre-trained model does not guarantee better accuracy. If its original training data differs significantly from your task, it may need substantial fine-tuning or may not be suitable.

**Key point:** Pre-trained models help us reuse existing knowledge and can reduce the time, data, and computation needed to build a new model.

## 4. Training from Scratch vs Pre-trained Models

There are two common approaches to building a deep learning model: training from scratch and using a pre-trained model.

### Training from Scratch

Training from scratch means creating a model with randomly initialized weights and training it using your own dataset.

The model starts without previously learned patterns and learns from the training data.

### Using a Pre-trained Model

A pre-trained model has already learned patterns from an earlier training task. We reuse its learned weights and adapt the model to our new task if needed.

### Comparison

| Factor | Training from Scratch | Pre-trained Model |
|---|---|---|
| Initial weights | Randomly initialized | Previously learned |
| Training time | Usually longer | Often shorter |
| Data requirement | Often needs more data | Can work with less data |
| Computing needs | Often higher | Often lower for adaptation |
| Initial knowledge | No learned patterns | Has learned patterns |
| Best suited for | Sufficient data and resources, or a specialized task | Many tasks related to the original training |

### Computer Vision Example

Suppose you want to classify images of defective and non-defective products.

**Approach 1 — Training from scratch:**
- Create a CNN architecture.
- Initialize its weights randomly.
- Train it using your product images.
- Evaluate its predictions.

**Approach 2 — Using a pre-trained model:**
- Load a model such as ResNet-18 with pre-trained weights.
- Replace or adapt its classification layer for two classes.
- Train the new classification layer and optionally fine-tune other layers.
- Evaluate its predictions.

The second approach often provides a useful starting point, especially when your dataset is small.

**Key point:** Training from scratch learns from an initial random state, while a pre-trained approach reuses patterns learned during previous training. Neither approach guarantees better results in every situation.

## 5. Model Weights

**Model weights** are numerical values inside a neural network that are learned and adjusted during training.

These values help the model recognize patterns and produce predictions.

During training, the model compares its predictions with the correct answers and updates its weights to reduce prediction errors.

### Example in Computer Vision

Suppose a CNN learns to classify images of cats and dogs.

During training, its weights are adjusted to help it recognize useful visual patterns, such as edges, textures, and shapes.

When a new image is given, the model uses its learned weights to calculate a prediction.

### Important points

- Weights are learned during training.
- They influence how the model processes its input.
- Pre-trained models contain weights learned from previous training.
- Saving and loading weights allows us to reuse a model's learned knowledge.

**Key point:** Model weights store learned numerical values that help a neural network make predictions.

## 6. Model Parameters

**Model parameters** are the values inside a neural network that can be learned and updated during training.

The two main types of learnable parameters are:

- **Weights:** Control how strongly input values influence the calculations.
- **Biases:** Add an extra value to a calculation, helping the model adjust its output.

### Example

Imagine a simple neural network calculation:

Output = (Input × Weight) + Bias

Suppose:

- Input = 5
- Weight = 2
- Bias = 1

The output is:

(5 × 2) + 1 = 11

During training, the model adjusts its weight and bias to improve its predictions.

### Parameters vs Weights

Weights are one type of model parameter. Biases are another type.

For example, a neural network may contain thousands or millions of parameters, depending on its architecture.

**Key point:** Model parameters are the learnable values in a neural network. Weights and biases are the main examples.

```python
import torch
import torch.nn as nn

# Create a layer with 3 inputs and 2 outputs
layer = nn.Linear(3, 2)

# Display weights and biases
print("Weights:")
print(layer.weight)

print("\nBiases:")
print(layer.bias)

# Count learnable parameters
total_parameters = sum(
    parameter.numel()
    for parameter in layer.parameters()
)

print("\nTotal parameters:", total_parameters)
```

Why 8?
- Weights: \(3 \times 2 = 6\)
- Biases: \(2\)
- Total: \(6 + 2 = 8\)
The actual weight and bias values will vary because PyTorch initializes them automatically.

## 7. Checkpoints

A **checkpoint** is a saved snapshot of a model's state at a particular point during training or after training.

It allows us to save progress and load it later instead of starting again from the beginning.

A checkpoint may contain:
- Model parameters, including weights and biases.
- Optimizer state, which helps resume training.
- The training epoch or step.
- Other information needed to continue training.

### Example

Suppose you are training a CNN for 20 epochs.

After epoch 10, you save a checkpoint. Later, if training is interrupted, you can load that checkpoint and continue training from the saved state rather than restarting from epoch 1.

### Model weights vs Checkpoint

- **Model weights:** The learned numerical values used by the model.
- **Checkpoint:** A saved snapshot that can include model weights, optimizer state, and training progress.

A checkpoint does not necessarily contain all these items; its contents depend on what was saved.

**Key point:** A checkpoint preserves model state and, when configured appropriately, training progress so that work can be resumed later.

```python
import torch
import torch.nn as nn
import torch.optim as optim

model = nn.Linear(3, 2)
optimizer = optim.Adam(model.parameters())

# Save a checkpoint
checkpoint = {
    "epoch": 5,
    "model_state": model.state_dict(),
    "optimizer_state": optimizer.state_dict()
}

torch.save(checkpoint, "model_checkpoint.pth")

print("Checkpoint saved")
```

```python
checkpoint = torch.load(
    "model_checkpoint.pth",
    weights_only=True
)

model.load_state_dict(checkpoint["model_state"])
optimizer.load_state_dict(checkpoint["optimizer_state"])

start_epoch = checkpoint["epoch"] + 1

print("Resume from epoch:", start_epoch)
```

This example demonstrates saving and restoring state. To resume a real training run correctly, you would also restore the training data pipeline and any other relevant training state.

## 8. Model Architecture

**Model architecture** is the structure or design of a neural network. It defines the layers, how they are connected, and how data flows through the model.

The architecture determines how the model processes input data and produces output.

For example, ResNet-18 is a model architecture. A pre-trained ResNet-18 has the same basic architecture but contains weights learned during training.

If you save only a model's weights, you generally need the matching architecture definition to load and use them correctly.

**Key point:** Architecture defines how a model is built, while weights determine what the model has learned.

## 9. Inference

**Inference** is the process of using a trained model to make predictions on new input data.

During inference, the model uses its learned weights to process the input and produce an output. It does not normally update its weights.

### Example in Computer Vision

Suppose you have a pre-trained model that identifies objects in images.

You provide a new image of a car.

The model processes the image and produces a prediction, such as:

- Input: Image of a car
- Model: Pre-trained image classification model
- Output: Car

### What happens during inference?

1. Load the trained model.
2. Prepare the input in the required format.
3. Pass the input through the model.
4. Obtain the prediction.
5. Interpret the output.

Inference is used in applications such as image classification, object detection, image segmentation, and face recognition.

**Key point:** Inference means using an already-trained model to make predictions. It is different from training because the model's weights are not normally updated.

```python
import torch
from torchvision import models, transforms
from PIL import Image

# Load a pre-trained model
model = models.resnet18(weights="DEFAULT")
model.eval()

# Prepare an image
image = Image.open(r"C:\Users\nanda\OneDrive\Pictures\Image\car.jpg").convert("RGB")

preprocess = transforms.Compose([
    transforms.Resize(256),
    transforms.CenterCrop(224),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

input_tensor = preprocess(image).unsqueeze(0)

# Run inference
with torch.no_grad():
    output = model(input_tensor)

predicted_class = output.argmax(dim=1).item()

print("Predicted class index:", predicted_class)
```

## 10. Training vs Inference

**Training** and **inference** are two different stages of working with a machine learning model.

### Training

Training is the process in which a model learns patterns from a dataset by adjusting its weights to improve predictions.

- Input: Training data and correct answers (labels).
- Process: Make predictions, calculate the loss, and update model weights.
- Output: A trained model.

### Inference

Inference is the process of using the trained model to make predictions on new data.

- Input: New images or other data.
- Process: Use the learned weights to calculate predictions.
- Output: Predicted classes, scores, or other results.

### Comparison

| Factor | Training | Inference |
|---|---|---|
| Purpose | Learn patterns | Make predictions |
| Data | Training dataset | New input data |
| Weights | Updated during training | Normally unchanged |
| Loss calculation | Used to guide learning | Not required to make predictions |
| Backpropagation | Used | Not normally used |
| Computation | Often more expensive | Often less expensive per prediction |

### Computer Vision Example

Imagine building a model to classify cats and dogs.

**During training:**
- The model receives labeled cat and dog images.
- It makes predictions and calculates the loss.
- Its weights are updated to improve its predictions.

**During inference:**
- The trained model receives a new image.
- It uses its learned weights.
- It predicts whether the image contains a cat or a dog.

**Key point: Training changes the model's learned weights; inference uses those learned weights to make predictions without normally changing them.**

```python
import torch
import torch.nn as nn

model = nn.Linear(2, 1)

# TRAINING
model.train()

input_data = torch.tensor([[2.0, 3.0]])
target = torch.tensor([[1.0]])

optimizer = torch.optim.SGD(model.parameters(), lr=0.01)
loss_function = nn.MSELoss()

prediction = model(input_data)
loss = loss_function(prediction, target)

optimizer.zero_grad()
loss.backward()
optimizer.step()

print("Training loss:", loss.item())

# INFERENCE
model.eval()

with torch.no_grad():
    new_input = torch.tensor([[4.0, 5.0]])
    result = model(new_input)

print("Inference prediction:", result.item())
```

## 11. Fine-tuning Introduction

**Fine-tuning** means taking a pre-trained model and continuing its training on a new dataset so that it becomes better suited to a specific task.

The model already has learned patterns from its original training. Fine-tuning adapts those learned patterns to the new task by updating some or all of its weights.

### Example in Computer Vision

Suppose a ResNet-18 model was pre-trained on ImageNet, a dataset containing many types of objects.

Now, you want to identify defective and non-defective products in factory images.

You can:

1. Load the pre-trained ResNet-18 model.
2. Replace its final classification layer with one suitable for your two classes.
3. Train the new layer and optionally update other layers.
4. Evaluate the fine-tuned model on unseen product images.

### Two common approaches

**1. Freeze most of the model**

Keep the pre-trained layers unchanged and train the new classification layer.

This is often a good starting point when the new dataset is small.

**2. Unfreeze and fine-tune layers**

Allow some or all pre-trained layers to update during training.

This can help the model adapt more closely to the new dataset, but it requires careful training to avoid damaging useful learned features.

### Fine-tuning vs training from scratch

- **Fine-tuning:** Starts with previously learned weights.
- **Training from scratch:** Starts with randomly initialized weights.

**Key point:** Fine-tuning adapts a pre-trained model to a new task by continuing training and updating selected weights.

```python
import torch.nn as nn
from torchvision import models

# Load a pre-trained model
model = models.resnet18(weights="DEFAULT")

# Freeze existing parameters
for parameter in model.parameters():
    parameter.requires_grad = False

# Replace the final classification layer
model.fc = nn.Linear(
    model.fc.in_features,
    2
)

# Check which parameters can be trained
for name, parameter in model.named_parameters():
    if parameter.requires_grad:
        print(name)
```

## 12. Transfer Learning

**Transfer learning** is a technique in which a model reuses knowledge learned from one task to help solve another task.

Instead of learning everything from the beginning, we take a pre-trained model and adapt it to our new problem.

### Example in Computer Vision

Suppose a model has already learned to recognize objects from a large image dataset.

Now, you want to classify images of different types of industrial products.

You can reuse the pre-trained model's learned visual features and adapt its classification layer to recognize your product categories.

### How transfer learning works

1. Load a pre-trained model.
2. Reuse its learned features.
3. Replace or adapt the final layer for the new task.
4. Train the new layer, and optionally fine-tune other layers.
5. Evaluate the model on new images.

### Transfer Learning vs Fine-tuning

- **Transfer learning** is the broader approach of reusing knowledge from a previously trained model for a new task.
- **Fine-tuning** is one way to apply transfer learning by continuing training and updating some or all of the pre-trained weights.

Transfer learning can also be performed by freezing the pre-trained layers and training only a new classification layer.

**Key point:** Transfer learning reuses knowledge from an existing model. Fine-tuning is one method of adapting that model by updating its weights.

```python
import torch.nn as nn
from torchvision import models

# Load a pre-trained model
model = models.resnet18(weights="DEFAULT")

# Reuse the pre-trained feature extractor
# Replace the original output layer for 3 new classes
model.fc = nn.Linear(model.fc.in_features, 3)

print(model.fc)
```

## 13. Model Input

**Model input** is the data given to a model so that it can make a prediction.

In Computer Vision, the input is usually an image. However, a model cannot always accept an image file directly. The image must be converted into the format and dimensions expected by the model.

### What does a pre-trained image model expect?

For example, a standard pre-trained ResNet-18 model expects an image represented as a tensor with:

- **Channels:** 3 for RGB images.
- **Height:** 224 pixels.
- **Width:** 224 pixels.
- **Pixel values:** Converted to floating-point values and normalized appropriately.

For a single image, the input shape is:

`[1, 3, 224, 224]`

Here:
- `1` = number of images in the batch.
- `3` = RGB color channels.
- `224` = image height.
- `224` = image width.

### Input preparation workflow

Image file → Resize and crop → Convert to tensor → Normalize → Model input.

The exact preprocessing requirements depend on the model and its training configuration. Using incorrect preprocessing can reduce prediction quality.

**Key point:** Model input is the data supplied to a model in the specific format that the model expects.

```python
from PIL import Image
from torchvision import models
from torchvision.models import ResNet18_Weights

# Load the pre-trained model and its recommended preprocessing
weights = ResNet18_Weights.DEFAULT
model = models.resnet18(weights=weights)
model.eval()

preprocess = weights.transforms()

# Open your image
image = Image.open(r"C:\Users\nanda\OneDrive\Pictures\Image\car.jpg").convert("RGB")

# Prepare the model input
input_tensor = preprocess(image).unsqueeze(0)

print("Input shape:", input_tensor.shape)
```

## 14. Model Output

**Model output** is the result produced by a model after processing its input.

In Computer Vision, the output depends on the task the model was trained to perform.

### Types of model output

**1. Image Classification**

The model predicts a class for the entire image.

Example:
- Input: Image of a car
- Output: Car

**2. Object Detection**

The model identifies objects and predicts their locations in the image.

Example:
- Input: Image containing cars and people
- Output: Class labels, bounding boxes, and confidence scores.

**3. Image Segmentation**

The model predicts a class for individual pixels or produces object masks.

Example:
- Input: Image of a road
- Output: A mask identifying road pixels.

### Raw output vs final prediction

A model does not always directly return a human-readable class name.

For example, a classification model might produce three scores:

- Cat: 2.1
- Dog: 4.8
- Rabbit: 1.3

The highest score corresponds to Dog. These raw scores are called **logits**. They are not probabilities unless an appropriate transformation is applied.

**Key point:** Model output is the result generated by the model. Its format depends on the task, such as classification, detection, or segmentation.

```python
import torch
from PIL import Image
from torchvision.models import resnet18, ResNet18_Weights

# Load the model and its recommended preprocessing
weights = ResNet18_Weights.DEFAULT
model = resnet18(weights=weights)
model.eval()

# Prepare an image
image = Image.open(r"C:\Users\nanda\OneDrive\Pictures\Image\car.jpg").convert("RGB")
input_tensor = weights.transforms()(image).unsqueeze(0)

# Generate model output
with torch.no_grad():
    logits = model(input_tensor)

# Convert scores to probabilities
probabilities = torch.softmax(logits, dim=1)

# Find the predicted class
class_index = probabilities.argmax(dim=1).item()
class_name = weights.meta["categories"][class_index]
confidence = probabilities[0, class_index].item()

print("Predicted class:", class_name)
print("Score:", round(confidence, 4))
```

## 15. Model Limitations

A **model limitation** is a weakness or restriction that affects how well a model performs or where it can be used reliably.

Even a pre-trained model that performs well on its original dataset may not perform well on every new task.

### Common limitations

**1. Dataset differences**

A model trained on general images may struggle with industrial images that look very different from its original training data.

**2. Limited class knowledge**

A pre-trained classification model can predict only among the classes it was trained to recognize. It will not automatically recognize new custom categories.

**3. Incorrect predictions**

A model can misclassify an image because of poor lighting, blur, unusual angles, or visually similar objects.

**4. Computing requirements**

Large models may require considerable memory and processing power, especially for high-resolution images or real-time applications.

**5. Bias in training data**

If the original dataset does not represent the images encountered in real use, the model may perform poorly on those images.

**6. Preprocessing requirements**

Incorrect image resizing, normalization, or input format can reduce prediction quality.

### Computer Vision example

Suppose a pre-trained model recognizes ordinary cars well but is used to identify tiny defects on metal surfaces.

The model may fail because:
- Small defects were not represented adequately in its original training.
- The defects require fine visual details.
- Its existing classes do not include the required defect categories.

You may need a suitable dataset, a new classification layer, fine-tuning, or a different model.

**Key point:** A pre-trained model is not suitable for every task automatically. Its performance depends on the data, task, architecture, and how it is used.

## 16. Pre-trained Model Workflow

The workflow **Dataset → Training → Model Weights → Pre-trained Model → Inference** explains how a model is trained and then used to make predictions.

### Step 1: Dataset

A dataset contains the examples used to teach a model.

For image classification, it contains images and their corresponding labels.

Example: Images labeled as cars, bikes, and buses.

### Step 2: Training

The model learns patterns from the training dataset. During training, it adjusts its weights to improve its predictions.

### Step 3: Model Weights

The learned weights are the numerical values adjusted during training. They store patterns that the model can reuse.

### Step 4: Pre-trained Model

When the trained model is saved and made available for reuse, it can be loaded as a pre-trained model for another task or for prediction.

The model architecture and its learned weights are both needed to use the model correctly.

### Step 5: Inference

The model receives a new image and uses its learned weights to generate a prediction.

Example: A new image is provided, and the model predicts that it contains a bus.

### Complete workflow

Dataset → Training → Learned Weights → Saved Trained Model → Inference → Prediction

**Important:** A model becomes pre-trained relative to a later task because it has already undergone training. The same model can be fine-tuned for a different task and then used for inference again.

## 17. Training ≠ Inference

Training and inference are different stages of using a machine learning model. They have different purposes.

### Training

Training teaches the model to recognize patterns from data.

- Uses training data.
- Calculates prediction errors using a loss function.
- Uses backpropagation to calculate gradients.
- Updates model weights using an optimizer.
- Produces a trained model.

### Inference

Inference uses the trained model to make predictions on new input data.

- Receives new input data.
- Uses the model's learned weights.
- Produces predictions.
- Normally does not update the model weights.
- Does not require backpropagation.

### Computer Vision example

Imagine building a model to identify cars.

**During training:**

The model receives labeled car images, makes predictions, calculates the loss, and updates its weights to improve its performance.

**During inference:**

You give the trained model a new car image. The model processes it using its learned weights and predicts the class.

### Comparison

| Training | Inference |
|---|---|
| Learns patterns | Uses learned patterns |
| Updates weights | Normally keeps weights unchanged |
| Uses a loss and optimizer | Does not need an optimizer |
| Uses backpropagation | Normally does not use backpropagation |
| Requires training data and labels for supervised learning | Can predict without knowing the correct label |

**Key point: Training teaches the model; inference uses what the model has learned.**

```python
import torch
import torch.nn as nn

# Create a simple model
model = nn.Linear(1, 1)

# Training data: y = 2x
X_train = torch.tensor([[1.0], [2.0], [3.0]])
y_train = torch.tensor([[2.0], [4.0], [6.0]])

loss_function = nn.MSELoss()
optimizer = torch.optim.SGD(model.parameters(), lr=0.01)

# TRAINING
model.train()

for epoch in range(500):
    predictions = model(X_train)
    loss = loss_function(predictions, y_train)

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

print("Training completed")

# INFERENCE
model.eval()

with torch.no_grad():
    new_input = torch.tensor([[4.0]])
    prediction = model(new_input)

print("Input:", new_input.item())
print("Predicted output:", round(prediction.item(), 2))
```
