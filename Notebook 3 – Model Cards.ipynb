# Model 3: ResNet-50
### 1. Model Name

The third model selected for this analysis is **ResNet-50**, a pretrained deep learning model used for computer vision.

Its official model name is `microsoft/resnet-50`.

Unlike BERT and DistilBERT, which process text, ResNet-50 processes images. It is commonly used to classify images into categories, such as identifying whether an image contains a dog, a car, or a bird.

The model uses a pretrained set of weights to recognize visual patterns, so we can use it without training the entire network from scratch.

### 2. Model Architecture

ResNet-50 uses a **Convolutional Neural Network (CNN)** architecture designed to recognize patterns in images.

Its key feature is **residual connections**, also called skip connections. These connections allow information to bypass some layers, helping very deep networks learn more effectively.

ResNet-50 contains 50 layers in its commonly described architecture and approximately 25.6 million parameters.

For example, when processing a car image, early layers can identify edges and textures, while deeper layers learn more complex patterns such as wheels and vehicle shapes.

**Why is this important?**

Residual connections help train deeper neural networks and make ResNet-50 useful for image-classification tasks. <Cite refs={["turn985857search0"]}/>

### 3. Intended Use

ResNet-50 is designed primarily for **image classification**, where a model predicts which category an image belongs to.

For example, given an image of a car, ResNet-50 can predict a category such as “sports car” or “convertible,” depending on the classes it was trained to recognize.

The pretrained ImageNet version can classify images into 1,000 categories. It can also be adapted to other image-related tasks through transfer learning and fine-tuning.

However, the standard classification model does not automatically draw bounding boxes around objects or identify every object in an image.

**Why is this important?**

Understanding the intended use helps us decide whether ResNet-50 is suitable for image classification or whether we need a different model for tasks such as object detection or segmentation.

### 4. Training Data

**Training data** refers to the images used to teach a model to recognize visual patterns.

The standard pretrained ResNet-50 model is commonly trained on **ImageNet**, a large dataset containing images organized into different categories.

The ImageNet-1K version uses 1,000 image categories, including animals, vehicles, and everyday objects.

By learning from these images, ResNet-50 develops the ability to recognize visual features and classify new images.

However, its predictions may be less reliable for objects or image types that differ significantly from its training data.

**Why is this important?**

Knowing the training data helps us understand which types of images the model is familiar with and whether it may need fine-tuning for a new task.

### 5. Input Format

ResNet-50 accepts **images converted into numerical tensors** rather than ordinary text.

Before an image is passed to the model, it must be resized to the expected dimensions and converted into a tensor. The standard pretrained ImageNet version typically uses images resized and cropped to 224 × 224 pixels, with three color channels (RGB).

The image values are also normalized using the preprocessing expected by the pretrained model.

For example, a photograph of a car is converted into a numerical representation that ResNet-50 can process.

**Why is this important?**

Correct image preprocessing helps the model interpret the input properly and produce reliable predictions.

```python
from transformers import AutoImageProcessor
from PIL import Image

model_name = "microsoft/resnet-50"

processor = AutoImageProcessor.from_pretrained(model_name)

image = Image.open(r"C:\Users\nanda\OneDrive\Pictures\Image\car.jpg").convert("RGB")
inputs = processor(images=image, return_tensors="pt")

print("Input tensor shape:", inputs["pixel_values"].shape)
```

### 6. Output Format

ResNet-50 returns numerical scores representing how strongly the model associates an image with each of its learned categories.

For the standard ImageNet-1K classification model, the output contains 1,000 scores, one for each category.

The category with the highest score is usually selected as the predicted class. These raw scores are called **logits**; they are not probabilities until an appropriate conversion, such as softmax, is applied.

For example, when given a car image, the model may assign a high score to a vehicle category.

**Why is this important?**

Understanding the output helps us interpret the model's predictions and identify the category it considers most likely.

```python
import torch
from transformers import AutoImageProcessor, ResNetForImageClassification
from PIL import Image

model_name = "microsoft/resnet-50"

processor = AutoImageProcessor.from_pretrained(model_name)
model = ResNetForImageClassification.from_pretrained(model_name)

image = Image.open(r"C:\Users\nanda\OneDrive\Pictures\Image\car.jpg").convert("RGB")
inputs = processor(images=image, return_tensors="pt")

model.eval()

with torch.no_grad():
    outputs = model(**inputs)

predicted_id = outputs.logits.argmax(-1).item()
predicted_label = model.config.id2label[predicted_id]

print("Predicted category:", predicted_label)
print("Number of class scores:", outputs.logits.shape[-1])
```

### 7. Limitations

ResNet-50 is effective for image classification, but it has limitations.

- **Fixed categories:** The standard pretrained model predicts among its 1,000 ImageNet categories. It may not recognize a specialized object that is absent from those categories.
- **Image quality:** Blurry, dark, or low-resolution images can reduce prediction accuracy.
- **Unfamiliar images:** Images that differ significantly from the training data may produce incorrect predictions.
- **Classification only:** The standard model identifies an image category but does not automatically locate objects with bounding boxes or produce segmentation masks.

For example, a model may classify a car image incorrectly if the vehicle is heavily obscured or belongs to a category it was not trained to recognize.

**Why is this important?**

Understanding these limitations helps us decide whether ResNet-50 is suitable for our image-classification task or whether we need a different model or additional training.

For example, a model trained mainly on clear photographs of vehicles may struggle with vehicles captured at night or in unusual weather conditions.

### 11. Example Usage

ResNet-50 can be used to classify an image by loading its pretrained weights, preparing the image, and passing it through the model.

**Why is this important?**

```python
import torch
from PIL import Image
from transformers import AutoImageProcessor, ResNetForImageClassification

model_name = "microsoft/resnet-50"

processor = AutoImageProcessor.from_pretrained(model_name)
model = ResNetForImageClassification.from_pretrained(model_name)

image = Image.open("car.jpg").convert("RGB")
inputs = processor(images=image, return_tensors="pt")

model.eval()

with torch.no_grad():
    outputs = model(**inputs)

predicted_id = outputs.logits.argmax(-1).item()
predicted_label = model.config.id2label[predicted_id]

print("Predicted category:", predicted_label)
```
