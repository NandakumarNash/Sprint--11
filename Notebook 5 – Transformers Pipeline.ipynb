# Topic 1 – Introduction to Hugging Face Pipelines

## 1. What is a Pipeline?

A **pipeline** in Hugging Face Transformers is a simple interface that allows us to use a pre-trained model for a particular task without manually writing every processing step.

For example, we can use a pipeline to:
- Identify whether a sentence is positive or negative.
- Identify names of people and organizations in text.
- Generate new text.
- Classify objects in an image.
- Convert speech into text.

## 2. Why Do We Use Pipelines?

Without a pipeline, we may need to load a tokenizer or image processor, convert the input into tensors, run the model, and convert the predictions into readable results.

A pipeline handles these steps for us.

## 3. Basic Syntax

`pipeline()` is imported from the `transformers` library.

We specify the task and optionally the model we want to use.

Example syntax:

`pipeline("sentiment-analysis")`

This creates a pipeline for sentiment analysis using a suitable default model.

## 4. What Happens Internally?

A pipeline generally performs these steps:

1. **Input:** Receives text, an image, or audio.
2. **Preprocessing:** Converts the input into the format the model expects. For text, this usually involves tokenization.
3. **Model inference:** The pre-trained model processes the input and produces predictions.
4. **Postprocessing:** Converts model outputs into understandable results, such as labels and scores.

## 5. Important Point

A pipeline does not automatically train a new model. It normally uses a pre-trained model to perform inference. You can specify a different compatible model for the task.

## 6. Limitations

- The first run may download model files.
- Large models require more memory and processing time.
- Predictions can be incorrect or biased.
- Results depend on the selected model and its training data.
- A pipeline simplifies usage, but it does not remove the need to understand the model's limitations.

```python
from transformers import pipeline

classifier = pipeline("sentiment-analysis")

text = "I really enjoyed learning about transformers."

result = classifier(text)

print(result)
```

Understanding the output
- label: The predicted sentiment category.
- score: The model's confidence score for that predicted label.
- 0.99: Approximately 99% model confidence, not a guarantee that the prediction is correct.
The first execution may take time because Hugging Face may need to download the model and its supporting files.

# Topic 2 – Sentiment Analysis

## 1. Task
Sentiment analysis identifies whether a piece of text expresses a positive or negative sentiment.

## 2. Model
We will use a pre-trained Hugging Face sentiment analysis model through the `pipeline()` function.

## 3. Input
Example sentence:

"I really enjoyed this movie."

## 4. Processing
1. The pipeline receives the sentence.
2. The tokenizer converts the text into tokens and numerical IDs.
3. The model processes the tokens and predicts sentiment.
4. The pipeline converts the prediction into a label and score.

## 5. Output
The output contains:
- **label:** Predicted sentiment, such as `POSITIVE` or `NEGATIVE`.
- **score:** The model's confidence score for the predicted label.

## 6. Interpretation
If the output label is `POSITIVE`, the model predicts that the sentence expresses positive sentiment.

A higher score indicates greater model confidence, but it does not guarantee that the prediction is correct.

## 7. Limitations
- Sarcasm and jokes can be difficult to identify.
- The model may misunderstand context.
- The model may not perform well on every language or domain.
- Sentiment labels depend on the model's training data.

```python
from transformers import pipeline

# Create a sentiment analysis pipeline
sentiment_analyzer = pipeline("sentiment-analysis")

# Input text
text = "I really enjoyed this movie."

# Perform sentiment analysis
result = sentiment_analyzer(text)

# Display the result
print(result)
```

# Topic 3 – Text Classification

## 1. Task
Text classification assigns a category or label to a piece of text.

For example, an email can be classified as spam or not spam, or a news article can be classified as sports, business, or politics.

## 2. Model
We will use a pre-trained Hugging Face text classification model through the `pipeline()` function.

## 3. Input
Example sentence:

"The team won the cricket match yesterday."

## 4. Processing
1. The pipeline receives the input text.
2. The tokenizer converts the text into numerical token IDs.
3. The model processes the tokens and predicts a category.
4. The pipeline converts the prediction into a readable label and score.

## 5. Output
The output generally contains:
- **label:** The predicted category.
- **score:** The model's confidence score for that prediction.

The available labels depend on the selected model.

## 6. Interpretation
A text classification model predicts which category best matches the input according to its training.

Text classification is a general task. Sentiment analysis is one specific type of text classification that focuses on sentiment, such as positive or negative.

## 7. Limitations
- Predictions depend on the model's available categories.
- A model trained for one classification task may not work correctly for another.
- Ambiguous text may be classified incorrectly.
- Confidence scores do not guarantee correctness.

```python
from transformers import pipeline

# Create a text classification pipeline
classifier = pipeline(
    "text-classification",
    model="ProsusAI/finbert"
)

# Input text
text = "The company reported strong profits this quarter."

# Perform text classification
result = classifier(text)

# Display the result
print(result)
```

Important: ProsusAI/finbert is trained for financial sentiment classification, so its labels relate to financial sentiment rather than general news categories.

# Topic 4 – Named Entity Recognition (NER)

## 1. Task
Named Entity Recognition (NER) identifies important entities in text, such as:
- Person names
- Organizations
- Locations

For example, in the sentence "Sundar Pichai works at Google in California," NER identifies the person, organization, and location.

## 2. Model
We will use a pre-trained Hugging Face token-classification model through the `pipeline()` function.

Model: `dslim/bert-base-NER`

## 3. Input
Example sentence:

"Satya Nadella works at Microsoft in the United States."

## 4. Processing
1. The pipeline receives the input sentence.
2. The tokenizer splits the sentence into tokens.
3. The model predicts an entity label for each relevant token.
4. The pipeline groups related tokens into complete entities where possible.
5. The result includes entity text, entity label, and confidence score.

## 5. Output
The output may identify:
- `Satya Nadella` as a person (`PER`)
- `Microsoft` as an organization (`ORG`)
- `United States` as a location (`LOC`)

The exact labels and scores depend on the model's predictions.

## 6. Interpretation
NER helps computers identify who or what is mentioned in a sentence.

For example, it can extract people, companies, and places from documents without manually reading every sentence.

## 7. Limitations
- Names can be ambiguous.
- Unfamiliar names may be misclassified.
- Some entities may be missed.
- Performance depends on the model's training data and language.
- The model may identify entity boundaries incorrectly.

```python
from transformers import pipeline

# Create a named entity recognition pipeline
ner = pipeline(
    "ner",
    model="dslim/bert-base-NER",
    aggregation_strategy="simple"
)

# Input text
text = "Satya Nadella works at Microsoft in the United States."

# Identify named entities
result = ner(text)

# Display the result
print(result)
```

The model may also identify United States as a location.
Important: NER models may use labels such as PER, ORG, and LOC. The aggregation_strategy="simple" setting groups related token predictions into entity spans.

# Topic 5 – Text Generation

## 1. Task
Text generation is the process of creating new text based on an input prompt.

For example, a model can complete a sentence, generate a paragraph, or continue a story.

## 2. Model
We will use a pre-trained text generation model through the Hugging Face `pipeline()` function.

Model: `gpt2`

## 3. Input
Example prompt:

"Artificial intelligence is useful because"

## 4. Processing
1. The pipeline receives the input prompt.
2. The tokenizer converts the prompt into numerical token IDs.
3. The model predicts possible next tokens based on the preceding text.
4. The generation process selects tokens and continues until a stopping condition is reached.
5. The pipeline converts the generated tokens back into readable text.

## 5. Output
The output contains text generated from the original prompt.

Example:

"Artificial intelligence is useful because it can help people solve problems and automate repetitive tasks."

This is an illustrative example. The actual output may be different.

## 6. Interpretation
The model generates text by predicting likely next tokens based on patterns learned during training.

Unlike classification, which assigns labels to input text, text generation produces new text.

## 7. Limitations
- Generated information may be incorrect.
- The model may produce repetitive or irrelevant text.
- Results depend on the prompt and generation settings.
- The model may reflect biases in its training data.
- Generated text should be checked before being used as factual information.

```python
from transformers import pipeline

# Create a text generation pipeline
generator = pipeline(
    "text-generation",
    model="gpt2"
)

# Input prompt
prompt = "Artificial intelligence is useful because"

# Generate text
result = generator(
    prompt,
    max_new_tokens=40,
    num_return_sequences=1
)

# Display the generated text
print(result[0]["generated_text"])
```

Understanding the parameters:
- max_new_tokens=40: Generates up to 40 new tokens beyond the prompt.
- num_return_sequences=1: Returns one generated sequence.
- generated_text: Contains the original prompt followed by the generated continuation.
The exact output will vary. The first run may download the model.

# Topic 6 – Image Classification

## 1. Task
Image classification identifies the main object or category present in an image.

For example, a model may classify an image as a cat, dog, car, or bird.

## 2. Model
We will use a pre-trained image classification model through the Hugging Face `pipeline()` function.

Model: `google/vit-base-patch16-224`

ViT stands for Vision Transformer.

## 3. Input
An image containing an object, such as a cat or dog.

The image can be provided as a local image file or a supported image URL.

## 4. Processing
1. The pipeline receives the image.
2. The image processor resizes and normalizes the image as required by the model.
3. The image is converted into a format the model can process.
4. The model predicts scores for its learned image categories.
5. The pipeline returns the predicted labels and scores.

## 5. Output
The output generally contains:
- **label:** The predicted image category.
- **score:** The model's confidence score for that category.

The pipeline can return multiple candidate categories, ranked by score.

## 6. Interpretation
If the model predicts `golden retriever`, it considers that category a likely match for the image.

The prediction depends on the categories the model learned during training.

## 7. Limitations
- The model may misclassify unclear or unusual images.
- Poor lighting and image quality can affect predictions.
- The model may not recognize categories outside its learned label set.
- A high score does not guarantee that a prediction is correct.
- The model classifies the image but does not automatically locate the object within it.

```python
from transformers import pipeline

# Create an image classification pipeline
classifier = pipeline(
    "image-classification",
    model="google/vit-base-patch16-224"
)

# Path to your image
image_path = "sample.jpg"

# Classify the image
result = classifier(image_path, top_k=3)

# Display the predictions
for prediction in result:
    print(prediction["label"], prediction["score"])
```

# Topic 7 – Image-to-Text

## 1. Task
Image-to-text generates a natural-language description of an image.

This task is also called **image captioning**.

For example, a model may describe an image as "A brown dog playing with a ball in a park."

## 2. Model
We will use a pre-trained image captioning model through the Hugging Face `pipeline()` function.

Model: `Salesforce/blip-image-captioning-base`

BLIP stands for Bootstrapping Language-Image Pre-training.

## 3. Input
An image containing objects, people, animals, or a scene.

The image can be provided as a local image file or a supported image URL.

## 4. Processing
1. The pipeline receives the image.
2. The image processor prepares the image for the model.
3. The model analyzes visual information in the image.
4. The model generates words to describe the image.
5. The pipeline returns the generated caption as readable text.

## 5. Output
Example output:

"A dog is playing with a ball in a grassy field."

This is an illustrative example. The actual caption depends on the image and model.

## 6. Interpretation
The model converts visual information into a text description.

Image classification predicts a category, while image-to-text generates a descriptive sentence about the image.

## 7. Limitations
- The model may overlook small objects.
- It may incorrectly describe objects or activities.
- It may miss important background details.
- The generated caption may be incomplete.
- It cannot guarantee that every detail in the caption is correct.

```python
from transformers import pipeline

# Create an image-to-text pipeline
captioner = pipeline(
    "image-to-text",
    model="Salesforce/blip-image-captioning-base"
)

# Path to the input image
image_path = "sample.jpg"

# Generate an image caption
result = captioner(image_path)

# Display the caption
print(result[0]["generated_text"])
```

# Topic 8 – Speech Recognition

## 1. Task
Speech recognition converts spoken language in an audio recording into written text.

This task is also called **Automatic Speech Recognition (ASR)**.

For example, a recording of someone saying "Good morning, how are you?" can be converted into the text:

"Good morning, how are you?"

## 2. Model
We will use a pre-trained speech recognition model through the Hugging Face `pipeline()` function.

Model: `openai/whisper-tiny`

Whisper is a speech recognition model that can transcribe audio into text.

## 3. Input
An audio recording containing spoken words.

For example, a WAV audio file containing an English sentence.

## 4. Processing
1. The pipeline receives the audio file.
2. The audio is loaded and prepared in the format expected by the model.
3. The model analyzes the audio and predicts the spoken words.
4. The pipeline converts the predictions into readable text.
5. The transcription is returned as the output.

## 5. Output
Example output:

"Hello, welcome to the training session."

This is an illustrative example. The actual transcription depends on the audio recording.

## 6. Interpretation
Speech recognition allows computers to convert spoken communication into text.

It is used in voice assistants, meeting transcription, subtitles, and voice-controlled applications.

## 7. Limitations
- Background noise can affect transcription.
- Accents and pronunciation differences may cause errors.
- Multiple people speaking at once can be difficult to transcribe.
- Technical terms and unfamiliar names may be recognized incorrectly.
- The model may omit or incorrectly transcribe words.

```python
from transformers import pipeline

# Create a speech recognition pipeline
transcriber = pipeline(
    "automatic-speech-recognition",
    model="openai/whisper-tiny"
)

# Path to the audio file
audio_path = "sample.wav"

# Transcribe the audio
result = transcriber(audio_path)

# Display the recognized text
print(result["text"])
```

# Topic 9 – Understanding the Internal Pipeline Workflow

## 1. Task
Understand what happens internally when a Hugging Face pipeline receives input and produces a prediction.

A pipeline connects the input-processing components, pre-trained model, and output-processing steps into one simple interface.

## 2. Model
The components used depend on the selected task.

For example, a sentiment analysis pipeline uses:
- A tokenizer
- A pre-trained text classification model
- A postprocessing step to produce labels and scores

## 3. Input
Example input:

"I love learning artificial intelligence."

This text is passed to a sentiment analysis pipeline.

## 4. Processing

### Step 1 – Model Loading
The pipeline loads the selected pre-trained model and its required tokenizer or processor.

The model files may be downloaded the first time they are needed.

### Step 2 – Preprocessing
The tokenizer converts the input text into tokens and numerical IDs that the model can understand.

For image and audio tasks, the corresponding processor prepares the image or audio input instead.

### Step 3 – Model Inference
The prepared input is passed to the pre-trained model.

The model performs calculations using its learned weights and produces predictions.

### Step 4 – Postprocessing
The pipeline converts the model's raw outputs into a more understandable format, such as sentiment labels and scores.

### Step 5 – Final Output
The pipeline returns the result to the user.

## 5. Output
Example output:

`[{'label': 'POSITIVE', 'score': 0.99}]`

This is an illustrative result. The actual score depends on the model and input.

## 6. Interpretation
The pipeline simplifies several operations into a single function call.

For example:

`classifier("I love learning artificial intelligence.")`

Internally, the pipeline coordinates preprocessing, inference, and postprocessing to return a readable prediction.

## 7. Limitations
- Loading large models can require significant memory.
- The first run may take longer because model files need to be downloaded.
- Different tasks require compatible models and processors.
- A pipeline does not guarantee correct predictions.
- Advanced customization may require working directly with the tokenizer, processor, or model.

```python
from transformers import pipeline

# Step 1: Load the pipeline
classifier = pipeline("sentiment-analysis")

# Step 2: Provide input
text = "I love learning artificial intelligence."

# Step 3: Pipeline performs preprocessing,
# model inference, and postprocessing
result = classifier(text)

# Step 4: Display the final output
print(result)
```
