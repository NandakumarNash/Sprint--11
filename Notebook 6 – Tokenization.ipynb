# Topic 1 – What Is a Token?

## Simple Concept

A **token** is a small unit of text that a language model processes.

A token can be:
- A complete word
- Part of a word
- A punctuation mark
- A special symbol

For example, the sentence:

`I love learning AI!`

A simple word-level tokenizer might split it into:

`["I", "love", "learning", "AI", "!"]`

Here, each item is a token.

However, modern language models may split the same sentence differently depending on their tokenizer.

## Important Explanation

Computers do not directly process sentences as human-readable text. Language models typically process numerical representations of tokens.

The process begins by splitting text into tokens. Later, each token is mapped to a numerical ID.

For example:

Original text → Tokens → Token IDs

`"I love AI"`

→ `["I", "love", "AI"]`

→ `[12, 45, 87]`

*Note: These numbers are illustrative examples, not actual IDs from a specific model.*

Tokenization is the first important step in preparing text for a language model.

```python
text = "I love learning AI!"

# Simple word-level tokenization
tokens = text.split()

print("Original Text:", text)
print("Tokens:", tokens)
print("Number of Tokens:", len(tokens))
```

# Topic 2 – Why Tokenization Is Required

## Simple Concept

Tokenization converts text into smaller units that a language model can process.

Humans understand words and sentences directly, but language models work with numerical representations of text.

Tokenization helps prepare text for this process.

## Important Explanation

Tokenization is required for three main reasons:

**1. Break text into smaller units**

A sentence is divided into tokens so the model can process its parts.

Example:

`I love AI`

↓

`["I", "love", "AI"]`

**2. Convert tokens into numbers**

A language model uses token IDs instead of directly processing text characters.

Example:

`["I", "love", "AI"]`

↓

`[12, 45, 87]`

These IDs are illustrative; actual IDs depend on the tokenizer.

**3. Prepare text for model input**

Tokenization makes it possible to prepare text using token IDs, attention masks, padding, and truncation when required.

These steps help organize text into the format expected by a language model.

## Key Point

Tokenization connects human-readable text with the numerical input used by language models.

The basic process is:

Original Text → Tokens → Token IDs → Model Input

```python
text = "I love learning AI"

# Step 1: Split text into tokens
tokens = text.split()

# Step 2: Create illustrative token IDs
vocabulary = {
    "I": 1,
    "love": 2,
    "learning": 3,
    "AI": 4
}

token_ids = [vocabulary[token] for token in tokens]

print("Original Text:", text)
print("Tokens:", tokens)
print("Token IDs:", token_ids)
```

# Topic 3 – Word-Level Tokenization

## Simple Concept

Word-level tokenization splits a sentence into individual words.

Each word is treated as a separate token.

For example:

`"I love machine learning"`

↓

`["I", "love", "machine", "learning"]`

## Important Explanation

**1. How it works**

The tokenizer separates text at word boundaries, usually using spaces and additional rules for punctuation.

**2. Vocabulary**

Each unique word needs an entry in the vocabulary. The tokenizer uses that vocabulary to map words to token IDs.

**3. Main limitation**

Word-level tokenization can struggle with words that are not present in its vocabulary.

For example, if the vocabulary contains `"learn"` but not `"learning"`, the complete word `"learning"` may be treated as an unknown token.

**4. Why modern models use other approaches**

Word-level tokenization can create a very large vocabulary because different word forms may require separate entries. Subword tokenization helps address this limitation by splitting some words into smaller pieces.

## Key Point

Word-level tokenization is easy to understand, but it may struggle with unfamiliar words and large vocabularies.

```python
text = "I love machine learning!"

# Basic word-level tokenization
tokens = text.split()

print("Original Text:", text)
print("Tokens:", tokens)
print("Number of Tokens:", len(tokens))
```

```python
import re

text = "I love machine learning!"

tokens = re.findall(r"\w+|[^\w\s]", text)

print("Tokens:", tokens)
```

Remember: This is a basic word-level demonstration, not the exact tokenization method used by a pretrained language model.

# Topic 4 – Subword Tokenization

## Simple Concept

Subword tokenization splits words into smaller meaningful or frequently occurring pieces when needed.

For example, a tokenizer might split:

`"unhappiness"`

into:

`["un", "happi", "ness"]`

This is an illustrative example. The actual pieces depend on the tokenizer and its vocabulary.

## Important Explanation

**1. Why subword tokenization is used**

Word-level tokenization may not recognize unfamiliar words. Subword tokenization allows a model to represent many unfamiliar words using smaller units.

**2. How it works**

A tokenizer learns or uses a vocabulary of common words and subword pieces. It divides input text into pieces according to its tokenization algorithm.

**3. Main advantage**

It can represent many different words without requiring every complete word to have its own vocabulary entry.

**4. Where it is used**

Many modern language models use subword-based tokenization methods, such as BPE, WordPiece, or Unigram.

Different methods can split the same sentence into different tokens.

## Key Point

Word-level tokenization generally treats each word as a unit, while subword tokenization can divide a word into smaller pieces. This helps language models handle a wider range of text.

```python
from transformers import AutoTokenizer

model_name = "bert-base-uncased"

tokenizer = AutoTokenizer.from_pretrained(model_name)

text = "I love learning about tokenization"

tokens = tokenizer.tokenize(text)

print("Original Text:", text)
print("Subword Tokens:", tokens)
```

The exact token pieces depend on the tokenizer version, but BERT commonly uses ## to mark a subword that continues a preceding piece.

# Topic 5 – Token IDs

## Simple Concept

A **token ID** is a numerical value assigned to a token in a tokenizer's vocabulary.

Language models use these numbers to represent text as input.

For example:

Original Text:

`I love AI`

Tokens:

`["I", "love", "AI"]`

Token IDs:

`[12, 45, 87]`

*These numbers are illustrative examples, not actual IDs from a pretrained model.*

## Important Explanation

**1. Why token IDs are required**

Neural networks perform computations using numerical data. Token IDs provide numerical representations of the tokens in the input text.

**2. How token IDs are assigned**

Each token has a corresponding ID in the tokenizer's vocabulary. The same tokenizer generally maps the same token to the same ID.

**3. Token IDs depend on the tokenizer**

Different models may use different vocabularies and token IDs for the same text. Therefore, token IDs from one tokenizer should not be assumed to work with another model.

**4. What happens next**

The model uses token IDs to look up learned vector representations called embeddings. These vectors are then processed by the model.

## Key Point

Token IDs are the numerical representations of tokens. They allow text to be passed into a language model in a form it can process.

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")

text = "I love learning AI"

tokens = tokenizer.tokenize(text)
token_ids = tokenizer.convert_tokens_to_ids(tokens)

print("Original Text:", text)
print("Tokens:", tokens)
print("Token IDs:", token_ids)
```

The IDs shown are the expected values for the standard BERT uncased vocabulary.
Observe: Each token has a corresponding numerical ID. The IDs are vocabulary indexes, not numerical scores that indicate how important a word is.

# Topic 6 – Vocabulary

## Simple Concept

A **vocabulary** is the collection of tokens that a tokenizer knows.

Each token in the vocabulary has a unique token ID.

For example, a small vocabulary might look like this:

| Token | Token ID |
|---|---:|
| I | 0 |
| love | 1 |
| AI | 2 |
| learning | 3 |

This is a simplified example, not a real pretrained model vocabulary.

## Important Explanation

**1. How vocabulary works**

When text is tokenized, the tokenizer identifies tokens from its vocabulary and converts them into their corresponding IDs.

**2. Vocabulary size**

Vocabulary size is the total number of token entries available to the tokenizer.

A vocabulary can contain complete words, subword pieces, punctuation marks, and special tokens.

**3. What happens with unknown words?**

Depending on the tokenizer, an unfamiliar word may be divided into known subword pieces. If it cannot represent the text using those pieces, it may use an unknown token such as `[UNK]`.

**4. Does every model use the same vocabulary?**

No. Different models can have different vocabularies and token IDs. Use the tokenizer associated with the model you are working with.

## Key Point

Vocabulary stores the tokens and their corresponding IDs. Tokenization uses this vocabulary to convert text into numerical input for a language model.

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")

# Display vocabulary size
print("Vocabulary Size:", tokenizer.vocab_size)

# Display IDs for some tokens
words = ["hello", "world", "learning"]

for word in words:
    token_id = tokenizer.convert_tokens_to_ids(word)
    print(word, "->", token_id)
```

Observe: BERT's uncased vocabulary contains 30,522 entries. The vocabulary size and IDs can differ for other tokenizers.

# Topic 7 – Special Tokens

## Simple Concept

**Special tokens** are predefined tokens that help a language model understand the structure or purpose of text.

Unlike ordinary tokens, they have specific roles in model processing.

For example, BERT uses special tokens such as:

- `[CLS]` — placed at the beginning of a sequence in many BERT tasks.
- `[SEP]` — separates sequences or marks the end of a sequence.
- `[PAD]` — used to fill shorter sequences to a common length.
- `[UNK]` — represents a token that cannot be handled by the tokenizer's vocabulary.

## Important Explanation

**1. Why special tokens are required**

They provide structural information that ordinary text tokens may not communicate.

**2. How they are added**

A tokenizer can automatically add special tokens when converting text into model input. The exact tokens depend on the model.

**3. Are special tokens included in token IDs?**

Yes. Special tokens have their own IDs in the tokenizer's vocabulary.

**4. Do all models use the same special tokens?**

No. BERT uses tokens such as `[CLS]` and `[SEP]`, while other models may use different tokens for sequence boundaries, instructions, or other purposes.

## Key Point

Special tokens help a model identify important boundaries and structures in its input. Their meaning depends on the model and its training setup.

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")

text = "I love AI"

# Add the model's required special tokens
encoded = tokenizer(text, add_special_tokens=True)

print("Tokens:", tokenizer.convert_ids_to_tokens(encoded["input_ids"]))
print("Token IDs:", encoded["input_ids"])
print("Special Tokens:", tokenizer.special_tokens_map)
```

Observe: [CLS] and [SEP] are added automatically by the BERT tokenizer in this example. The [MASK] token is available in the vocabulary but is not automatically inserted into this sentence.

# Topic 8 – Attention Mask

## Simple Concept

An **attention mask** tells a language model which token positions should be attended to and which positions should be ignored as padding.

In a typical Hugging Face tokenizer output:

- `1` means the position contains a real token.
- `0` means the position is padding and should be masked out.

## Important Explanation

**1. Why an attention mask is required**

When multiple sentences have different lengths, padding may be added to make their sequences the same length. The attention mask helps the model distinguish real tokens from padding.

**2. How it works**

Suppose the input contains:

Token IDs: `[101, 1045, 102]`

Attention Mask: `[1, 1, 1]`

All three positions contain actual tokens, including the special tokens.

If padding is added:

Token IDs: `[101, 1045, 102, 0, 0]`

Attention Mask: `[1, 1, 1, 0, 0]`

The last two positions are padding and are marked with `0`.

**3. Important distinction**

An attention mask does not indicate whether a word is important or unimportant. It identifies positions to include or mask according to the model's attention rules.

## Key Point

Token IDs tell the model which tokens are present. The attention mask indicates which token positions should be attended to, especially when padding is used.

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")

texts = [
    "I love AI",
    "I love learning about artificial intelligence"
]

encoded = tokenizer(
    texts,
    padding=True,
    return_attention_mask=True
)

print("Token IDs:")
print(encoded["input_ids"])

print("\nAttention Masks:")
print(encoded["attention_mask"])

print("\nTokens:")
for ids in encoded["input_ids"]:
    print(tokenizer.convert_ids_to_tokens(ids))
```

Observe: Both sentences have the same sequence length. The shorter sentence has padding tokens with attention-mask values of 0.

# Topic 9 – Padding

## Simple Concept

**Padding** means adding special padding tokens to shorter sequences so that multiple sequences have the same length.

For example:

Sentence 1: `I love AI`

Sentence 2: `I love learning about AI`

Their token sequences have different lengths. Padding makes them equal in length when required.

## Important Explanation

**1. Why padding is required**

When processing multiple sentences together in a batch, models commonly need the sequences to have compatible dimensions. Padding allows shorter sequences to match the length of longer ones.

**2. How padding works**

Suppose the token IDs are:

Sentence 1: `[101, 1045, 2293, 9932, 102]`

Sentence 2: `[101, 1045, 2293, 4083, 102]`

If both already have the same length, no additional padding is needed.

If the first sentence has fewer tokens, the tokenizer can add `[PAD]` tokens until both sequences have the required length.

**3. Padding and attention masks**

Padding tokens are commonly assigned an attention-mask value of `0`, while actual tokens have a value of `1`.

**4. Padding strategies**

- `padding=True`: pad sequences to the longest sequence in the current batch.
- `padding="max_length"`: pad sequences to a specified maximum length.

## Key Point

Padding makes sequences the same length for batch processing. It does not add new meaning to the original sentence.

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")

texts = [
    "I love AI",
    "I love learning about artificial intelligence"
]

encoded = tokenizer(
    texts,
    padding=True,
    return_tensors="pt"
)

print("Token IDs:")
print(encoded["input_ids"])

print("\nAttention Masks:")
print(encoded["attention_mask"])

print("\nPadded Tokens:")
for ids in encoded["input_ids"]:
    print(tokenizer.convert_ids_to_tokens(ids))
```

# Topic 10 – Truncation

## Simple Concept

**Truncation** means removing tokens from a sequence when the sequence is longer than the allowed maximum length.

For example, suppose a tokenizer allows only five tokens:

Original tokens:

`["I", "love", "learning", "about", "artificial", "intelligence"]`

After truncation:

`["I", "love", "learning", "about", "artificial"]`

The remaining token is removed.

*This example ignores special tokens to keep the concept simple.*

## Important Explanation

**1. Why truncation is required**

Language models have input-length limits. Long text may need to be shortened before it can be processed.

**2. How truncation works**

The tokenizer limits the number of tokens according to the specified maximum length. Depending on the tokenizer and configuration, tokens can be removed from the beginning or the end.

**3. Padding vs truncation**

- Padding adds tokens to shorter sequences.
- Truncation removes tokens from longer sequences.

**4. Important consideration**

Truncation can remove useful information. Choose an appropriate maximum length for the task, and check whether important content is being lost.

## Key Point

Truncation controls sequence length by removing excess tokens. Padding and truncation are often used together to prepare batches of text.

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")

text = "I love learning about artificial intelligence and language models"

# Tokenize without truncation
original_tokens = tokenizer.tokenize(text)

# Tokenize with a maximum length of 6, including special tokens
encoded = tokenizer(
    text,
    truncation=True,
    max_length=6,
    add_special_tokens=True
)

truncated_tokens = tokenizer.convert_ids_to_tokens(
    encoded["input_ids"]
)

print("Original Tokens:", original_tokens)
print("Original Token Count:", len(original_tokens))

print("\nTruncated Tokens:", truncated_tokens)
print("Final Token Count:", len(encoded["input_ids"]))
```

# Topic 11 – Maximum Sequence Length

## Simple Concept

**Maximum sequence length** is the maximum number of token positions allowed for a model input in a particular configuration.

These positions can include ordinary tokens, special tokens, and padding tokens.

For example, if the maximum sequence length is `8`, the final input can contain up to eight token positions.

## Important Explanation

**1. Why it matters**

A maximum sequence length helps control how much text the model processes at once. Longer sequences generally require more memory and computation.

**2. Relationship with truncation**

If the text exceeds the configured maximum length, truncation can remove excess tokens.

**3. Relationship with padding**

If a sequence is shorter than the chosen fixed length, padding can fill the remaining positions.

**4. Special tokens count too**

For BERT, `[CLS]` and `[SEP]` usually occupy positions within the maximum length.

For example, with a maximum length of `8`, these two special tokens leave six positions for ordinary text tokens.

**5. Does every model have the same limit?**

No. Supported input lengths differ by model architecture and configuration. The tokenizer's settings and the model's actual limits must be considered together.

## Key Point

Maximum sequence length controls the number of token positions in the input. Truncation handles sequences that are too long, while padding can fill shorter sequences when a fixed length is needed.

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")

texts = [
    "I love AI",
    "I am learning natural language processing with transformers"
]

encoded = tokenizer(
    texts,
    padding="max_length",
    truncation=True,
    max_length=8,
    return_tensors="pt"
)

print("Token IDs:")
print(encoded["input_ids"])

print("\nAttention Masks:")
print(encoded["attention_mask"])

print("\nSequence Length:", encoded["input_ids"].shape[1])

print("\nTokens:")
for ids in encoded["input_ids"]:
    print(tokenizer.convert_ids_to_tokens(ids))
```

# Topic 12 – Complete Tokenization Experiment

## Objective

Understand how original text becomes model input through tokenization.

## Complete Process

**Stage 1: Original Text**

The sentence entered by the user.

Example: `"I love learning AI"`

**Stage 2: Tokens**

The tokenizer splits the sentence into tokens according to its own rules.

Example: `["i", "love", "learning", "ai"]`

**Stage 3: Token IDs**

Each token is mapped to its numerical ID from the tokenizer's vocabulary.

Example: `[1045, 2293, 4083, 9932]`

**Stage 4: Attention Mask**

The attention mask identifies token positions that should be attended to. For a single sentence without padding, the values are generally all `1`.

Example: `[1, 1, 1, 1]`

Special tokens may also be added, and padding or truncation can change the final sequence.

## Key Point

The tokenizer prepares text in the format expected by the model. The exact tokens and IDs depend on the tokenizer associated with that model.

```python
from transformers import AutoTokenizer

# Load the tokenizer
tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")

# Experiment with different sentences
texts = [
    "I love AI",
    "I am learning tokenization",
    "Transformers process natural language"
]

# Convert text into model inputs
encoded = tokenizer(
    texts,
    padding=True,
    truncation=True,
    max_length=12,
    return_attention_mask=True
)

# Display each stage
for i, text in enumerate(texts):

    tokens = tokenizer.convert_ids_to_tokens(
        encoded["input_ids"][i]
    )

    print("=" * 55)
    print("Original Text:", text)
    print("Tokens:", tokens)
    print("Token IDs:", encoded["input_ids"][i])
    print("Attention Mask:", encoded["attention_mask"][i])
```
